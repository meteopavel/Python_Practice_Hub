```python
################
### Счётчики ###
################

"""
Одна из популярных задач в программировании - что-нибудь посчитать:
программа в фитнес-тнекере может подсчитывать шаги пользователя, 
программное обеспечение лифта может определять его пробег, считая этажи
и остановки, программа проектировщика может посчитать, сколько на плане
помещений определённого размера.

Допустим, у нас есть квартира, а в ней несколько помещений, площадь этих
помещений положим в список, а затем посчитаем количество этих помещений:
"""
flat = [5.55, 22.19, 7.78, 26.86, 5.55,
        29.84, 22.19, 5.55, 16.85, 4.52
]
# len(flat) подсчитает количество элементов в списке flat
# Сохраним это значение в переменную rooms_num
rooms_num = len(flat)

# И напечатаем полученное значение
print(rooms_num)
```

```python
"""
Задача 1.
Теперь нам нужно выяснить, сколько в этом списке помещений, площадь
которых равна оперделённому значению. Например 22.19 м^2.

Для этого сделаем следующее:
- Создадим перемнную-счётчик, в ней будем хранить количество 
обнаруженных помещений нужной площади. В начале это значение - ноль.
- По очереди переберём каждое значение из списка, сравнивая его с 22.19
Для этого применим цикл for...in...
- Если очередное значение из списка равно 22.19 - увеличим значение 
переменной-счётчика на 1
- Когда все значения списка будут проверены - напечатаем сообщение:
Комнат площадью {искомая площадь} кв.м: {значение счётчика}
"""
flat = [5.55, 22.19, 7.78, 26.86, 5.55,
        29.84, 22.19, 5.55, 16.85, 4.52
]
room_size = 22.19
# Переменную для подсчёта чего-либо принято называть count
count = 0

for room in flat:
    if room == room_size:
        count += 1

print('Комнат площадью', room_size, 'кв.м:', count)
```

```python
"""
Добавление одного и того же значения к переменной (в цикле и в других
случаях) - это достаточно распространённая операция, для неё придумали 
специальный оператор +=. С его помощь код становится короче и проще.
"""
counter = 0
# Например, вместо строки:
counter = counter + 10
# можно написать:
counter += 10

# Подобный оператор есть и для вычитания:
counter -= 10

# И для умножения, деления и вообще арифметических операций:
counter *= 10
counter /= 10
counter %= 10    # деление с остатком
counter //= 10   # целочисленное деление
counter **= 10   # возведение в степень
```

```python
"""
Задача 2.
Сосчитать суммарную площадь всех помещений. Принцип решения будет чем-то
похож на подход в предыдущей задаче.

Для решения этой задачи никаких сравнений не понадобится; суммировать
нужно все значения списка, без исключений.

- Создаём переменную sum_area, к значению которой будем по очереди
прибавлять значения каждого из элементов списка. Пока что значение
этой переменной — ноль.
- В цикле перебираем все элементы списка flat, как и в прошлой задаче.
- Значение каждого элемента списка прибавляем к sum_area.
- Напечатаем сообщение, в котором укажем общую площадь квартиры.
"""
flat = [5.55, 22.19, 7.78, 26.86, 5.55,
        29.84, 22.19, 5.55, 16.85, 4.52
]
sum_area = 0

for area in flat:
    sum_area += area

print("Общая площадь квартиры:" , round(sum_area, 2), "кв.м.")
```

```python
"""
Задача 3.
Нужно посчитать, сколько альбомов выпустила музыкальная группа
Depeche Mode в 21 веке.
"""
# Список годов, в которые Depeche Mode выпускала альбомы
years = [
    1981, 1982, 1983, 1984, 1986, 1987, 1990,
    1993, 1997, 2001, 2005, 2009, 2013, 2017
]

count = 0

for year in years:
    if year >= 2000:
        count += 1

print('Выпущено альбомов в XXI веке:', count)
```

```python
"""
Задача 4.
Нужно модифицировать задачу 1. А именно, нужно вынести код для подсчёта
комнат определенной площади в отдельную функцию:
- Объявить функцию, назовите её rooms_equal()
- Функция rooms_equal() должна принимать на вход два параметра:
   - room_size — значение, которое функция будет искать в списке;
   - room_list — имя списка, в котором будет проводиться поиск;
- Перенести в функцию код, который подсчитывает количество помещений
заданной площади.
Вызвать функцию rooms_equal(), чтобы посчитать, сколько в списке flat
помещений площадью 5.55 кв.м.
Добавить ещё один вызов функции: пусть она сосчитает, сколько комнат
площадью 9.2 кв.м в списке hut (англ. hut — «хижина»).
"""
def rooms_equal(room_size, room_list):
    count = 0
    for room in room_list:
        if room == room_size:
            count += 1
    print('Комнат площадью', room_size, 'кв.м:', count)

flat = [
    5.55, 22.19, 7.78, 26.86, 5.55,
    29.84, 22.19, 5.55, 16.85, 4.52
]

hut = [9.2, 3.5, 8.1, 2.3, 9.2, 4.2, 6.9]

rooms_equal(5.55, flat)
rooms_equal(9.2, hut)
```

```python
"""
Задача 5.
Дописать функцию number_of_occurrences() так, чтобы она считала, сколько
раз в переданной строке (параметр string) встречается буква, переданная 
в параметр char.

Строка - это последовательность, как и список; строку можно точно так же
перебрать в цикле по элементам (буквам и другим символам), найти и
пересчитать нужные элементы.

Предыдущая задача очень похода на эту.
"""
def number_of_occurrences(char, string):
    count = 0
    for letter in string:
        if letter == char:
            count += 1
    print('Исходная строка:', string)
    print('Количество вхождений символа', char, 'составляет:', count)

phrase = ('Не волнуйтесь, если что-то не работает. Если бы всё'
          'работало, вас бы уволили.')

number_of_occurrences('е', phrase)
```
