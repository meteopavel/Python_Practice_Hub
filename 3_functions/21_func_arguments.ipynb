```python
#########################
### Аргументы функции ###
#########################

"""
Иногда говорят, "функция ожидает аргумент"; это значит, что при вызове
функции в неё должны быть переданы какие-то значения.

Если функцию вызвать, а аргумент по какой-то причине не передать, то
будет получена ошибка: 
TypeError: func_name missing 1 required positional argument 'arg_name'
"""
def print_home(name, planet):
    print(name + ' живёт на планете ' + planet)

print_home('Йода')
# Потерялся обязательный позиционный аргумент 'planet'
```

```python
"""
Чтобы "защититься" от таких ошибок, можно присвоить любому параметру
значение по умолчанию. Оно будет передано в тело функции, если при 
вызове ожидаемый аргумент не был получен.

Значение по умолчанию присваивается параметрам при объявлении функции: 
"""
def print_home(name='Инкогнито', planet='Икс'):
    print(name + ' живёт на планете ' + planet)

print_home('Дроид-убийца')
# Передаём только один аргумент вместо двух, но ошибки уже нет
```

```python
"""
При вызове функции print_home() значения передаются в параметры
соответственно с их позицией, в порядке их перечисления: первый
аргумент в первый параметр, а второй - во второй. Это называется
*позиционные аргументы*.

Это просто и удобно, но может вызвать путаницу. Например, если 
при вызове забыли передать первый аргумент, то аргумент, который
должен быть вторым пердастся в первый параметр!
"""
def print_home(name='Инкогнито', planet='Икс'):
    print(name + ' живёт на планете ' + planet)

print_home('Земля')
```

```python
"""
В предыдущем коде Python увидел, что в функцию передан лишь один аргумент
и решил, что это первый *позиционный* аргумент. Этот аргумент был передан
в параметр name - и в результате получилась полная чепуха.

Причина проста: Python не понял, какой именно аргумент был передан при
вызове функции.

Чтобы избежать такой неразберихи, при вызове функции можно передавать 
*именнованные* аргументы - явно указать, какому параметру какой аргумент
соответствует.

Приведём примеры такой передачи:
"""
def print_home(name='Инкогнито', planet='Икс'):
    print(name + ' живёт на планете ' + planet)

# Передадим именованный параметр:
# явно укажем, что значение "Земля" предназначено для параметра planet
print_home(planet='Земля')

# Ещё раз вызовем функцию: передадим два именованных параметра, 
# но не в том порядке, как они указаны в объявлении функции:
print_home(planet='Земля', name='Винни Пух')
print_home(planet='Марс', name='Марк Уотни')

# Если аргументы поименованы, то их можно передавать в любом порядке: 
# функция увидит имена и во всём сама разберётся
```

```python
"""
Задача.
Написать функцию под названием create_profile, которая принимает
следующие именованные аргументы:
- name: строка, обязательный аргумент;
- age: целое число, обязательный аргумент;
- city: строка, необязательный аргумент, по умолчанию 'Москва'
- hobby: строка, необязательный аргумент, по умолчанию 'Нет хобби'
Функция должна возвращать словарь с информацией о профиле.
Пример использования:
profile1 = create_profile(name='Алексей', age=30)
profile2 = create_profile(
    name='Мария', age=25, city='Иркутск', hobby='Чтение'
)

print(profile1)
Вывод: {'name': 'Алексей', 'age': 25, 'city': 'Москва', 'hobby': 'Нет хобби'}

Пример создания словаря:
profile = {'name': name, 'age': age ...}

Чтобы функция сама ничего не печатала, необходимо вернуть словарь с
помощью ключевого слова return. Например return profile
"""
def create_profile(name, age, city='Москва', hobby='Нет хобби'):
    profile = {
        'name': name,
        'age': age,
        'city': city,
        'hobby': hobby
    }
    return profile
profile1 = create_profile(name='Алексей', age=30)
profile2 = create_profile(name='Мария', age=25,
                          city='Иркутск', hobby='Чтение')

print(profile1)
print(profile2)
```
