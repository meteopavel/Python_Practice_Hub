```python
###################################
### Возврат значений из функции ###
###################################

"""
До сих пор мы писали функции, которые печатали результаты вычислений на
экран. Но обычно задача функции не сводится к печати: Вычисленное в
функции значение передаётся в код, и с этим значением производится ещё
какая-то работа.

Функция должна работать примерно так: в неё поступает "заготовка" - 
аргументы, переданные при вызове, а на выход фукция выдаёт какой-то 
результат, *возвращает* его в ту часть кода, откуда она была вызвана.
В дальнейшей работе программа применит значение, вернувшееся из функции,
для каких-то дальнеёших вычислений.

Возвращаемое значение указывается в теле функции после ключевого слова
*return*.

Напишем фунцию calc_square(), которая получает на вход два аргумента - 
длины сторон прямоугольника, вычисляет и возвращает его площадь:
"""
def calc_square(side_a, side_b):
    result = side_a * side_b
    return result

rectangle_area = calc_square(16, 9)

print(rectangle_area)
```

```python
"""
Напишем небольшую программу, которая вычислит и напечатает суммарную 
площадь трёх комнат квартиры, размеры который нам известны: 3х5, 4х6 
и 3х6 метров. Используем ту же функцию calc_square()

Площадь каждой комнаты получим отдельно и запишем в разные переменные.
Затем просуммируем и напечатаем результат:
"""
def calc_square(side_a, side_b):
    result = side_a * side_b
    return result

room1 = calc_square(3, 5)
room2 = calc_square(4, 6)
room3 = calc_square(3, 6)

rooms_sum = room1 + room2 + room3
print('Суммарная площадь комнат равна', rooms_sum, 'кв.м')
```

```python
"""
Задача 1.
Необходимо доработать функцию comfort_count() так, чтобы она 
*возвращала* посчитанное количество комфортных дней.
Необходимо присвоить возвращенное значение переменной nice_days, а затем
распечатать результат по аналогии с предыдущей ячейкой.
"""
may_2017 = [24, 26, 15, 10, 15, 19, 10, 1, 4, 7, 7, 7, 12, 14, 17, 8,
            9, 19, 21, 22, 11, 15, 19, 23, 15, 21, 16, 13, 25, 17, 19]

def comfort_count(temperatures):
    count = 0
    for temp in temperatures:
        if 22 <= temp <= 26:
            count += 1
    return count

nice_days = comfort_count(may_2017)

print('Количество тёплых дней в этом месяце:', nice_days)
```

```python
"""
Задача 2.
Дизайнер Афанасий решил построить из стекла и металлических уголков 
восемь одинаковых кубов: ребра кубов из уголков, а грани - из стекла.
Ребро куба должно быть 3 метра. 

Нужно написать программу, которая напечатает общую длину уголков,
необходимых для строительства восьми кубов.

Понадобится функция, которая будет принимать длину стороны куба, 
а возвращать длину всех его рёбер. Задачу выполнить по аналогии 
с предыдущим кодом. Имена функции и переменных придумать самостоятельно.
"""
def calc_cube_perimeter(side):
    return side * 12


one_cube_perimeter = calc_cube_perimeter(3)
full_length = one_cube_perimeter * 8

print('Длина всех рёбер для 8 кубов равна:', full_length)
```

```python
"""
Задача 3.
Теперь нужно сделать функцию для расчёта площади стекла для граней куба.
Функция для расчёта должна принимать длину стороны куба, а возвращать
площадь стекла. 
"""
def calc_cube_area(side):
    return side * side * 6


one_cube_area = calc_cube_area(3)
full_area = one_cube_area * 8

print('Площадь стекла для 8 кубов равна:', full_area)
```
