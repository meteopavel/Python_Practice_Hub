```python
###########################################
### f-строки: переменные прямо в тексте ###
###########################################

"""
Для составления строки из фрагментов текста и переменных мы можем
использовать конкатенацию: 
"""
weather = 'облачно'
print('На улице сейчас ' + weather + '.')
"""
Но более удобным способом являются f-строки.

Применение f-строк позволяет включать значения переменных прямо 
в текстовую строку, обрамлённую кавычками. Никаких символов +, 
никакой путаницы с кавычками. 

f-строки выглядят также, как и обычные строки, только перед открывающей
кавычкой ставится символ f:
"""
f'Это f-строка'
"""
Имена переменных в f-строках обрамляются в фигурные скобки {}:
"""
weather = 'облачно'
print(f'На улице сейчас {weather}.')
```

```python
"""
Форматирование через f-строки даёт ещё одно преимущество: если переменная
в f-строке имеет тип, отличный от str, то Python автоматически приведёт
тип этой переменной к строке.
"""
one_hundred = 100
rubles = 'рублей'
friends = 'друзей'
print(f'Не имей {one_hundred} {rubles}, а имей {one_hundred} {friends}.')
# Без f-строк код выглядит похуже:
print('Не имей ' + str(one_hundred) + ' ' + rubles + 
      ', а имей ' + str(one_hundred) + ' ' + friends + '.')
"""
f-строка лаконичнее и удобнее, разница в коде видна очень хорошо. Если
вспомнить Дзен Питтона "Простое лучше, чем сложное", то его смело можно
применить к f-строкам.
"""
```

```python
"""
Задача 1.
Заменить конкатенацию на f-строку:
print('Сейчас ' + weather + ', на градуснике ' + str(temperature) + '.')
Нужно создать функцию show_meteo, которая будет принимать температуру
и состояние погоды, а возвращать f-строку. Функцию нужно вызвать
три раза с разными аргументами
"""
def show_meteo(temperature, weather):
    return f'Сейчас {weather}, на градуснике {temperature}.'

print(show_meteo(24, 'солнечно'))
print(show_meteo(17, 'облачно'))
print(show_meteo(3, 'дождливо'))
```

```python
"""
Задача 2.
Написать функцию, которая принимает целое число, а возвращает его
факториал, при этом применить f-строку: Факториал X равен Y 
Вызвать три раза с разными аргументами. 
"""
def factorial(n):
    result = 1
    for k in range(2, n + 1):
        result *= k
    return f'Факториал {n} равен {result}'

print(factorial(5))
print(factorial(10))
print(factorial(15))
```
