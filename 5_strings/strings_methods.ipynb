```python
####################
### Методы строк ###
####################

"""
Методы строк в Python - это встроенные функции, которые позволяют 
выполнять различные операции с текстовыми данными. Строки в Python
явялются неизменяемыми, поэтому все методы строк возвращают новую
строку, а исходная строка остаётся неизменной.
"""
```

```python
"""
Методы изменения регистра:
1. str.upper() - преобразует все символы строки в верхний регистр
2. str.lower() - преобразует все символы строки в нижний регистр
3. str.capitalize() - делает первую букву строки заглавной,
                      а остальные - строчными
4. str.title() - делает первую букву каждого слова заглавной
5. str.swapcase() - меняет регистр всех символов на противоположный
"""
text = 'hello'
print(text.upper())  # HELLO

text = 'HELLO'
print(text.lower())  # hello

text = 'hello world'
print(text.capitalize())  # Hello world

text = 'hello world'
print(text.title())  # Hello World

text = 'Hello World'
print(text.swapcase())  # hELLO wORLD
```

```python
"""
Методы поиска и замены:
1. str.find(substring) - ищет подстроку substring в строке и возвращает
                         индекс первого вхождения. Если подстрока не
                         найдена, возвращает -1
2. str.index(substring) - аналогично find, но вызывает ошибку ValueError,
                          если подстрока не найдена.
3. str.replace(old, new) - заменяет все вхождения подстроки old на new
"""
text = 'hello world'
print(text.find('world'))  # 6
print(text.find('Python'))  # -1

text = 'hello world'
print(text.index('world'))  # 6
# print(text.index('Python'))  # ValueError

text = 'hello world'
print(text.replace('world', 'Python'))  # hello Python
```

```python
"""
Методы разделения и объединения строк
1. str.split(separator) - разделяет строку на список по указанному
                          разделителю
2. str.join(iterable) - объединяет элементы списка (или другого
                        итерируемого объекта) в строку, используя
                        строку как разделитель
"""
text = 'apple,banana,cherry'
print(text.split(','))  # ['apple', 'banana', 'cherry']

fruits = ['apple', 'banana', 'cherry']
print(','.join(fruits))  # 'apple,banana,cherry'
```

```python
"""
Методы удаления пробелов и специальных символов

1. str.strip() - удаляет пробельные символы (пробелы, табуляции
                 переводы строк) в начале и в конце строки
2. str.lstrip() - удаляет пробельные символы только в начале строки
3. str.rstrip() - удаляет пробельные символы только в конце строки
"""
text = '   hello   '
print(text.strip())  # 'hello'
print(text.lstrip())  # 'hello   '
print(text.rstrip())  # '   hello'
```

```python
"""
Методы проверки содержимого строки

1. str.startswith(prefix) - проверяет, начинается ли строка с 
                            указанного префикса
2. str.endswith(suffix) - проверяет, заканчивается ли строка
                          указанным суффуксом
3. str.isalpha() - проверяет, состоит ли строка только из букв
4. str.isdigit() - проверяет, состоит ли строка только из цифр
5. str.isalnum() - проверяет, состоит ли строка только из букв и цифр
6. str.isspace() - проверяет, состоит ли строка только из
                   пробельных символов
"""
text = 'hello world'
print(text.startswith('hello'))  # True
print(text.endswith('world'))  # True

print('hello'.isalpha())  # True
print('hello123'.isalpha())  # False

print('123'.isdigit())  # True
print('hello123'.isdigit())  # False

print('hello123'.isalnum())  # True
print('hello 123'.isalnum())  # False

print('   '.isspace())  # True
print('hello'.isspace())  # False
```

```python
"""
Методы форматирования и выравнивания

1. str.center(width, fillchar) - выравнивает строку по центру, 
                                 дополняя её символами fillchar
                                 до указанной длины
2. str.ljust(width, fillchar) - выравнивает строку по левому краю
3. str.rjust(width, fillchar) - выравнивает строку по правому краю
"""
text = 'hello'
print(text.center(10, '*'))  # **hello***
print(text.ljust(10, '-'))  # hello-----
print(text.rjust(10, '-'))  # -----hello
```

```python
"""
Методы кодировки и декодирования

1. str.encode(encoding) - кодирует строку в байты с использованием
                          указанной кодировки
2. str.decode(encoding) - декодирует байты обратно в строку
"""
text = 'hello'
print(text.encode('utf-8'))  # b'hello'

bytes_data = b'hello'
print(bytes_data.decode('utf-8'))  # 'hello'
```

```python
"""
Другие полезные методы

1. str.count(substring) - считает количество вхождений подстроки в строке
2. str.partition(separator) - разделяет строку на три части: 
                              до разделителя, сам разделитель и после
                              разделителя
3. str.zfill(witdh) - дополняет строку нулями до указанной длины 
"""
text = 'hello world hello'
print(text.count('hello'))  # 2

text = 'hello world'
print(text.partition(' '))  # ('hello', ' ', 'world')

text = '42'
print(text.zfill(5))  # 00042
```

```python
"""
Задача 1.

Написать программу, которая запрашивает у пользователя строку, 
а в ответ выводит эту строку:
    1. В верхнем регистре;
    2. В нижнем регистре;
    3. С заглавной буквы каждого слова

Пример работы программы:
Введите строку: hello world
Верхний регистр: HELLO WORLD
Нижний регистр: hello world
С заглавными буквами: Hello World
"""
user_input = input('Введите строку: ')

upper_case = user_input.upper()
lower_case = user_input.lower()
title_case = user_input.title()

print(f'Верхний регистр: {upper_case}')
print(f'Нижний регистр: {lower_case}')
print(f'С заглавными буквами: {title_case}')
```

```python
"""
Задача 2.

Написать программу, которая:
    1. Заправшивает у пользователя строку и подстроку;
    2. Проверяет, есть ли подстрока в строке;
    3. Если да, выводит индекс первого вхождения подстроки. 
    4. Если нет - сообщает об этом .

Пример работы программы:
Введите строку: hello world
Верхний подстроку: world
Подстрока найдена на позиции: 6

Введите строку: hello world
Верхний подстроку: python
Подстрока не найдена.
"""
main_string = input('Введите строку: ')
substring = input('Введите подстроку: ')
idx = main_string.find(substring)
if idx != -1:
    print(f'Подстрока найдена на позиции: {idx}')
else:
    print('Подстрока не найдена.')
```

```python
"""
Задача 10.

Написать программу, которая:
    1. Запрашивает у пользователя строку;
    2. Проверяет:
       *  Начинается ли строка с "http".
       *  Заканчивается ли строка на ".com".
    3. Выводит результат.
 
Пример работы программы:
Введите строку: http://example.com
Строка начинается с 'http': True
Строка заканчивается на '.com': True
"""
input_string = input('Введите строку: ')

starts_with_http = input_string.startswith('http')
ends_with_com = input_string.endswith('.com')

print(f'Строка начинается с "http": {starts_with_http}')
print(f'Строка заканчивается на ".com": {ends_with_com}')
```

```python
"""
Бонусное задание: Шифр Цезаря.

Написать программу, которая:
    1. Запрашивает у пользователя строку и число n;
    2. Сдвигает каждый символ строки на n позиций в алфавите (шифр Цезаря);
    3. Выводит зашифрованную строку.
 
Пример работы программы:
Введите строку: abc
Введите сдвиг: 3
Зашифрованная строка: def
"""
text = input('Введите строку: ')
shift = int(input('Введите сдвиг: '))

encrypted_text = []
for char in text:
    encrypted_text.append(chr(ord(char) + shift))

print(f'Зашифрованная строка: {"".join(encrypted_text)}')
```
