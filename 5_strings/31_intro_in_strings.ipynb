```python
#######################################
### Строка: последовательность букв ###
#######################################

"""
Тип данных "строка" - это коллекция, элементами которой служат символы,
составляющие строку.

Строку, как и любую коллекцию можно обрабатывать в цикле, извлекать из
неё элементы или группы элементов и преобразовывать в другие коллекции.

Строку можно преобразовать в список или сет:
"""
string = 'Я люблю кодить на языке Python'
new_list = list(string)
new_set = set(string)
print('Список из строки: ' + str(new_list))
print('Сет из строки: ' + str(new_set))
```

```python
"""
При создании списка из строки не было неожиданностей: получился список,
каждым элементом которого стали символы, включая пробелы и запятые.

При создании множества:
1. Из строки были исключены дублирующиеся символы;
2. При печати элементы вывелись в случайном порядке.
"""
```

```python
"""
Строка - это *упорядоченная* коллекция. К элементам строки можно
обратиться по индексу, точно так же, как и к элементам списка.
Например в строке 'Я люблю кодить на языке Python' под индексом [5]
будет получена буква 'л' 

Соберем индексы определённых букв в список и выведем эти буквы:
"""
monument_string = 'Я памятник себе воздвиг нерукотворный'

index_list = [0, 1, 2, 8, 6, 17, 24]

for i in index_list:
    print(monument_string[i])
```

```python
"""
Иногда бывает необходимо обратиться к элементам коллекции, начиная отсчёт
с конца. В таких ситуациях применяют отрицательные индексы: последний
элемент коллекции можно получить по индексу [-1], предпоследний [-2],
и так далее. 

Это касается строк и любых упорядоченных коллекций (например, списков):
"""
friends = ['Сергей', 'Соня', 'Миша', 'Дима', 'Алина']
print(friends[-3])  # Миша
print(friends[-5])  # Сергей

monument_string = 'Я памятник себе воздвиг нерукотворный'
print(monument_string[-2])   # ы
print(monument_string[-37])  # Я
```

```python
"""
Строку можно разделить на слова или на другие фрагмены текста - и далее
сохранить получившиеся подстроки в список. Для этого есть метод split()

Для метода split() можно указать аргумент - разделитель, по которому 
строка будет разбита на части. Разделителем может быть цифра, буква,
пробел или любой набор символов.

Например, если к строке 'молоковоз' применить метод split() с аргументом
'о' - то результатом будет список ['м', 'л', 'к', 'в', 'з']. При этом
сам разделитель вырезается из строки.
"""
milk_str = 'молоковоз'

new_list = milk_str.split('о')

print(new_list)
```

```python
counter_str = 'Раз-два-три-четыре-пять, вышел зайчик погулять'
# Преобразуем строку в список, а разделителем будет дефис
counter_list = counter_str.split('-')
print(counter_list)

blok_str = 'Ночь. Улица. Фонарь. Аптека'
# Разобьём фразу по словам, а разделителем будет сочетание точки и пробела
blok_list = blok_str.split('. ')
print(blok_list)
```

```python
"""
Чтобы разделить строку на слова - в качестве разделителя нужно
указать пробел, это такой же символ, как и буква или цифра.

Если аргумент метода split() пробел - то фраза, к которой применяют
этот метод, будет разбита на отдельные слова:
"""
poem_str = 'Дама сдавала  багаж'
words_list = poem_str.split(' ')
print(words_list)
words_list = poem_str.split()
print(words_list)
"""
Python нашёл в строке пробелы, воспринял их как разделители, разбил
строку на части и создал список из получившихся элементов; каждый
элемент этого списка - это строка, слово из исходной фразы.

Если вызвать split() *без аргуметов* - он точно также разобъёт строку
по пробелам.

Следовательно, чтобы разбить сроку на отдельные слова по пробелам - 
не обязательно указывать аргумент. Код words_list = poem_str.split(' ') 
и words_list = poem_str.split() дадут одинаковый результат. Но если 
между словами стоит более одного пробела, результат будет отличаться. 
В коде выше мы показали, что вызов split() без аргумента будет работать 
более корректно.
"""
```

```python
"""
Метод split() может использоваться, например, для поиска определённых
слов в тексте.

Текст можно преобразовать в список слов и проверить, есть ли определённое 
слово в этом списке, и если есть - выполнить нужные действия.

Например, в офис провайдера приходит сообщение от клиента:
"""
message = 'У меня всё сломалось и не работает соединение с интернетом!'

# Разобьём текст на слова
words = message.split()
# Проверим, есть ли ключевые слова в письме
if 'стоимость' in words:
    print('Переслать письмо в отдел биллинга')
elif 'сломалось' in words:
    print('Переслать письмо в техподдержку')
else:
    print('Содержание письма не определено')
```

```python
"""
Можно выполнить и обратную процедуру: сцепить строку из элементов
списка. В получившейся строке можно разделить элементы исходного
списка каким-нибудь текстовым символом или набором символов.

Эту операцию выполняет метод join().

Из элементов списка words_list создадим строку new_string и напечатаем
её.

Следует обратить внимание, что у метода join() синтаксис отличается
от split(): метод применяется к строке-разделителю, а список передается
как *аргумент* метода.

Разделителем может быть любая строка: какое-нибудь слово, запятая с 
пробелом ', ', пробел ' ', а может быть и пустая строка ''. 
"""
words_list = ['Раз', 'два', 'три', 'четыре', 'пять',
              'вышел', 'зайчик', 'погулять']

new_string = '-'.join(words_list)

print(new_string)
```

```python
"""
Задача 1.
Написать функцию, которая принимает аргументом любую фразу, а
возвращает третье с конца слово.

Функцию нужно вызвать по очереди с каждой фразой.
"""
quote_1 = 'Работает? Не трогай'
quote_2 = 'Если твой код работает, значит это хороший код'
quote_3 = 'Лень - главное достоинство программиста'

def return_word(phrase):
    words = phrase.split()
    
    if len(words) < 3:
        return 'Недостаточно слов'
    
    return words[-3]


print(return_word(quote_1))
print(return_word(quote_2))
print(return_word(quote_3))
```

```python
"""
Задача 2.

В коде подготовлен список запросов к Анфисе queries. Необходимо
определить, какие из них адресованы Аффисе, а какие - другим людям.

Нужно написать функцию check_query(), которая принимает запрос как 
параметр, анализирует его и возвращает одну из двух строк:

1. Если запрос начинается со слова 'Анфиса' - 'запрос к Анфисе'
2. Если запрос начинается с любого другого слова - 'запрос к кому-то ещё'
"""
queries = [
    'Анфиса, сколько у меня друзей?',
    'Андрей, ну где ты был?',
    'Андрей, ну обними меня скорей!',
    'Анфиса, кто все мои друзья?'
]

def check_query(query):
    name = query.split(' ')[0]
    if name == 'Анфиса,':
        return 'запрос к Анфисе'
    else:
        return 'запрос к кому-то ещё'
    
for query in queries:
    print(f'{query} - {check_query(query)}')
```

```python
"""
Задача 3. 

Теперь научим Анфису извлекать суть запроса. Нужно переписать код 
функции check_query() так, чтобы при любом запросе она "отрезала"
от строки имя и возвращала только запрос без имени.

Например:
Анфиса, сколько у меня друзей? — сколько у меня друзей?
"""
queries = [
    'Анфиса, сколько у меня друзей?',
    'Андрей, ну где ты был?',
    'Андрей, ну обними меня скорей!',
    'Анфиса, кто все мои друзья?'
]

def check_query(query):
    name = query.split(', ')[1]
    return f'{name}'

for query in queries:
    print(f'{query} - {check_query(query)}')
```
