```python
# Основные операции со списками #
"""
| Операция       | Пример              | Результат / действие          |
|----------------|---------------------|-------------------------------|
| Доступ по индексу | `fruits[0]`      | Возвращает первый элемент     |
| Обратный индекс   | `fruits[-1]`     | Возвращает последний элемент  |
| Срезы            | `fruits[1:]`      | Возвращает подсписок          |
| Изменение        | `fruits[1] = ...` | Меняет элемент списка         |
| Длина списка     | `len(fruits)`     | Возвращает длину списка       |
"""
```

```python
# Методы списков #
"""
1. .append(x)
Добавляет элемент x в конец списка.
fruits.append('дыня')
# ['яблоко', 'банан', 'вишня', 'дыня']

2. .extend(iterable)
Добавляет все элементы из другого итерируемого объекта (например, списка)
в конец.
fruits.extend(['дыня', 'виноград'])
# ['яблоко', 'банан', 'вишня', 'дыня', 'виноград']

3. .insert(i, x)
Вставляет элемент x на позицию i.
fruits.insert(1, 'груша')
# ['яблоко', 'груша', 'банан', 'вишня']

4. .remove(x)
Удаляет первое вхождение элемента x. Если элемент не найден — вызывает
ошибку ValueError.
fruits.remove('банан')
# ['яблоко', 'вишня']

5. .pop([i])
Удаляет и возвращает элемент по индексу i. Без аргумента удаляет
последний элемент.
last = fruits.pop()
# last = 'вишня'
# fruits = ['яблоко', 'банан']

6. .clear()
Очищает список.
fruits.clear()
# fruits = []

7. .index(x[, start[, end]])
Возвращает индекс первого элемента, равного x. Необязательно указывать
диапазон поиска.
letters = ['a', 'b', 'c']
print(letters.index('b'))  # 1

8. .count(x)
Возвращает количество раз, которое x встречается в списке.
nums = [1, 2, 2, 3]
print(nums.count(2))  # 2

9. .sort(key=None, reverse=False)
Сортирует список на месте.
nums = [3, 1, 4, 2]
nums.sort()
print(nums)  # [1, 2, 3, 4]
# Обратная сортировка
nums.sort(reverse=True)
print(nums)  # [4, 3, 2, 1]

10. .reverse()
Разворачивает список на месте.
nums = [1, 2, 3]
nums.reverse()
print(nums)  # [3, 2, 1]

11. .copy()
Возвращает поверхностную копию списка.
original = [1, 2, 3]
copy = original.copy()
copy.append(4)
print(original)  # [1, 2, 3]
print(copy)      # [1, 2, 3, 4]
⚠️ Важно: если вы просто присвоите new_list = old_list, то вы получите ссылку на тот же объект!
"""
```

```python
# Задача 5: Объединение списков без повторений #
"""
Условие:
Даны два списка. Объедини их в один, удалив повторяющиеся элементы.
Пример:
list1 = [1, 2, 3]
list2 = [3, 4, 5]
Результат:
[1, 2, 3, 4, 5]
Подсказка:
Объедини списки с помощью +, а затем воспользуйся множеством set()
для удаления дублей. Но чтобы сохранить порядок — лучше сделать через цикл.
"""
list1 = [1, 2, 3]
list2 = [3, 4, 5]

combined = list1 + list2
result = []
seen = set(combined)

for item in combined:
    if item not in result:
        result.append(item)


print(result)
print(seen)
```

```python
# Задача 6: Проверка на упорядоченность #
"""
Условие:
Напиши функцию, которая проверяет, является ли список
отсортированным по возрастанию .
Пример:
is_sorted([1, 2, 3])     # True
is_sorted([1, 3, 2])     # False
Подсказка:
Пройдись по списку от первого до предпоследнего элемента.
Если какой-то следующий элемент меньше текущего — верни False.
"""
def is_sorted(nums):
    for i in range(len(nums) - 1):
        if nums[i] > nums[i + 1]:
            return False
    return True

print(is_sorted([1, 2, 3]))
print(is_sorted([1, 3, 2]))
```

```python
# Задача 9: Сумма соседей #
"""
Условие:
Дан список целых чисел. Создай новый список, каждый элемент которого
равен сумме соседей соответствующего элемента исходного списка.
Примечание:
Для крайних элементов (первого и последнего) сосед считается только один.
Пример:
nums = [1, 2, 3, 4]
Результат:
[2, 4, 6, 3]  # объяснение: [2, 1+3=4, 2+4=6, 3]
Подсказка:
Используй цикл for i in range(len(nums)) и проверяй границы индексов.
"""
nums = [1, 2, 3, 4]
result = []
for i in range(len(nums)):
    if i == 0:
        result.append(nums[1])
    elif i == len(nums) - 1:
        result.append(nums[len(nums) - 2])
    else:
        result.append(nums[i - 1] + nums[i + 1])
print(result)
```

```python
# Задача 14: Проверка наличия дубликатов #
"""
Условие:
Проверь, содержит ли список дубликаты.
Пример:
lst = [1, 2, 3, 2]
Результат:
True
Подсказка:
Сравни длину списка с длиной множества, созданного из этого списка.
"""
def is_duplicate(lst):
    lst_set = set(lst)
    if len(lst_set) != len(lst):
        return True
    return False

lst = [1, 2, 3, 2]
print(is_duplicate(lst))
```

```python
# Задача 15: Проверка наличия дубликатов #
"""
Условие:
Напиши программу, которая вычисляет среднее арифметическое
всех элементов списка.
Пример:
nums = [1, 2, 3, 4, 5]
Результат:
3.0
Подсказка:
Используй sum() и len()
"""
def arithmetic_mean(nums):
    return sum(nums) / len(nums)

nums = [1, 2, 3, 4, 5]
print (arithmetic_mean(nums))
```

```python
# Задача 16: Удаление минимального элемента (без min() и .remove()) #
"""
Условие:
Напиши код, который удаляет первое вхождение минимального числа
из списка, не используя min() и .remove() .
Пример:
nums = [5, 2, 8, 2, 1]
Результат:
[5, 2, 8, 2]
Подсказка:
Сначала найди минимальное значение вручную через цикл,
затем найди его первый индекс и используй .pop(i).
"""
def remove_min(nums):
    min_value = nums[0]
    for num in nums:
        if num < min_value:
            min_value = num
    min_value_index = nums.index(min_value)
    nums.pop(min_value_index)
    return nums

nums = [5, 2, 8, 2, 1]
print (remove_min(nums))
```

```python
# Задача 17: Сдвиг элементов вправо #
"""
Условие:
Напиши функцию, которая сдвигает все элементы списка на N позиций
вправо. Если N больше длины списка — используй остаток от деления.
Пример:
lst = [1, 2, 3, 4, 5]
n = 6
Результат:
[4, 5, 1, 2, 3]
Подсказка:
Используй срезы: lst[-n:] + lst[:-n]. Но не забудь обработать
случай, когда n больше длины списка.
"""
def shift_right(lst, n):
    if n > len(lst):
        n %= len(lst)
    return lst[-n:] + lst[:-n]

lst = [1, 2, 3, 4, 5]
n = 3
print(shift_right(lst, n))
```

```python
# Задача 19: Проверка на палиндром (через список) #
"""
Условие:
Проверь, является ли строка палиндромом , используя список.
Пример:
text = "radar"
Результат:
True
Подсказка:
Преобразуй строку в список символов и сравни его с обратной версией.
"""
def is_palindrome(text):
    text_lst = list(text)
    return text_lst == text_lst[::-1]

text = 'radar'
print(is_palindrome(text))
```

```python
# Задача 20: Поиск пар с заданной суммой #
"""
Условие:
Найди все уникальные пары элементов в списке, которые в сумме
дают заданное число.
Пример:
nums = [1, 2, 3, 4, 5]
target = 5
Результат:
[(1, 4), (2, 3)]
Подсказка:
Можно использовать вложенный цикл или множество для проверки
наличия нужного второго числа.
"""
def find_pairs_for_sum(lst, target):
    result = []
    for i in range(len(lst) - 1):
        for j in range(i + 1, len(lst)):
            if lst[i] + lst[j] == target:
                result.append((lst[i], lst[j]))
    return result


nums = [1, 2, 3, 4, 5]
target = 5
print(find_pairs_for_sum(nums, target))
```

```python
# Задача 21: Группировка по длине строк #
"""
Условие:
Есть список строк. Создай словарь, где ключ — это длина строки,
а значение — список слов этой длины.
Пример:
words = ['apple', 'bat', 'car', 'ant', 'orange']
Результат:
{
    5: ['apple'],
    3: ['bat', 'car', 'ant'],
    6: ['orange']
}
Подсказка:
Проходи по списку, добавляя каждое слово в соответствующий
список в словаре. Используй .setdefault() или проверку
if key in dict
"""
def dict_from_words(words):
    result = {}
    for word in words:
        key = len(word)
        if key not in result:
            result[key] = []
        result[key].append(word)
    return result

words = ['apple', 'bat', 'car', 'ant', 'orange']
print(dict_from_words(words))
```

```python
# Задача 22: Объединение списков без повторений и сортировкой #
"""
Условие:
Объедини несколько списков в один, удалив дубликаты и
отсортировав результат.
Пример:
list1 = [3, 2, 1]
list2 = [2, 4, 5]
list3 = [5, 6, 3]
Результат:
[1, 2, 3, 4, 5, 6]
Подсказка:
Объедини списки, преобразуй в множество, затем обратно в
список и отсортируй.

Расширили функционал, добавив возможность использовать любое
количество списков.
Заменили метод extend на вложенные циклы с методом append, а
затем уже эту конструкцию заменили на списочное включение
"""
def combine_and_sort(*args):
    combined = [x for item in args for x in item]
    # combined = []
    # for item in args:
    #     for x in item:
    #         combined.append(x)
    unique = set(combined)
    result = list(unique)
    result.sort()
    return result

list1 = [3, 2, 1]
list2 = [2, 4, 5]
list3 = [5, 6, 3]
list4 = [7, 7, 8]

print(combine_and_sort(list1, list2, list3, list4))
```

```python
# Задача 23: Самый длинный подряд идущий подсписок #
"""
Условие:
Дан список целых чисел. Найди самый длинный подряд идущий подсписок,
состоящий из одинаковых элементов.
Пример:
nums = [1, 1, 2, 2, 2, 3, 3, 1, 1, 1, 1]
Результат:
[1, 1, 1, 1]
Подсказка:
Используй переменные для текущего значения и максимальной длины.
Проходи по списку, сравнивая с предыдущим элементом.
"""
def longest_equal_sublist(nums):
    num_set = set(nums)
    cur_len = 0
    max_len = 0
    values_max_lens = {}
    for num in num_set:
        for n in nums:
            if n == num:
                cur_len += 1
                max_len = max(max_len, cur_len)
            else:
                cur_len = 0
        values_max_lens[num] = max_len
        max_len = 0
    max_key = max(values_max_lens, key=values_max_lens.get)
    max_key_value = values_max_lens[max_key]
    return [max_key] * max_key_value

nums = [1, 1, 2, 2, 2, 3, 3, 1, 1, 1, 1]
print(longest_equal_sublist(nums))
```

```python
# Задача 24: Найти элемент, встречающийся ровно один раз #
"""
Условие:
Дан список целых чисел. Найди первый элемент, который встречается
ровно один раз.
Пример:
nums = [4, 3, 2, 4, 1, 3, 2]
Результат:
1
Подсказка:
Используй словарь для подсчёта частоты, затем пройди по списку
и найди первый элемент с частотой 1.
"""
def one_element(nums):
    frequency = {}
    for num in nums:
        frequency[num] = frequency.get(num, 0) + 1

    for num in nums:
        if frequency[num] == 1:
            return num

nums = [4, 3, 2, 4, 1, 3, 2]
print(one_element(nums))
```

```python
# Задача 25: Все элементы в порядке возрастания частоты #
"""
Условие:
Отсортируй список так, чтобы элементы шли по возрастанию количества их появления.
Пример:
items = ['a', 'b', 'c', 'a', 'b', 'a']
Результат:
['c', 'b', 'b', 'a', 'a', 'a']
Подсказка:
Используй sorted() с key=lambda x: frequency[x], где frequency — словарь частот.
"""
def sorted_frequency_list(letters):
    frequency = {}
    for letter in letters:
        frequency[letter] = frequency.get(letter, 0) + 1

    sorted_letters = sorted(letters, key=lambda x: frequency[x])
    return sorted_letters

letters = ['a', 'b', 'c', 'a', 'b', 'a']
print(sorted_frequency_list(letters))
```

```python
# Задача 26: Проверка, является ли список "пилообразным" #
"""
Условие:
Список называется пилообразным, если каждый элемент строго больше
или строго меньше своих соседей:
a[0] < a[1] > a[2] < a[3] > a[4] < ... или a[0] > a[1] < a[2] > a[3] < ...
Напиши функцию, которая проверяет, является ли список пилообразным.
Пример:
nums = [1, 3, 2, 4, 1]
Результат:
True  # потому что 1 < 3 > 2 < 4 > 1
Подсказка:
Проверь два возможных шаблона: начиная с < или с >.
"""
def is_list_sawtooth(lst):
    n = len(lst)
    if n <= 2:
        return len(set(lst)) == n
    return all((lst[i - 1] < lst[i] > lst[i + 1]) or (lst[i - 1] > lst[i] < lst[i + 1])
               for i in range(1, n - 1))

nums = [1, 3, 2, 4, 1]
print(is_list_sawtooth(nums))
```

```python
# Задача 27: Разделение на чётные и нечётные #
"""
Условие:
Переставь элементы списка так, чтобы сначала шли все чётные, а потом — все нечётные,
сохранив их относительный порядок.
Пример:
nums = [1, 2, 3, 4, 5, 6]
Результат:
[2, 4, 6, 1, 3, 5]
Подсказка:
Можно использовать два списка: один для чётных, другой для нечётных, потом объединить.
"""
def swapped_even_odd(nums):
    even = [x for x in nums if x % 2 == 0]
    odd = [x for x in nums if x % 2 != 0]
    return even + odd

nums = [1, 2, 3, 4, 5, 6]
print(swapped_even_odd(nums))
```

```python
# Задача 28: Поиск ближайшего числа #
"""
Условие:
Дан список чисел и число x. Найди элемент списка, который наиболее близок к x.
Пример:
nums = [10, 3, 8, 15]
x = 7
Результат:
8
"""
def find_closest(nums, x):
    return min(nums, key=lambda num: abs(num - x))

nums = [10, 3, 8, 15]
x = 7
print(find_closest(nums, x))
```

```python
# Задача 29: Все пары с разностью K #
"""
Условие:
Найди все уникальные пары чисел из списка, у которых разность по модулю равна K.
Пример:
nums = [1, 3, 5, 7]
k = 2
Результат:
[(1, 3), (3, 5), (5, 7)]
"""
def unique_pairs(nums, k):
    pairs = []
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if abs(nums[i] - nums[j]) == k:
                pairs.append((nums[i], nums[j]))
    return pairs

nums = [1, 3, 5, 7]
k = 2
print(unique_pairs(nums, k))
```

```python
# Задача 30: Группировка по первому символу #
"""
Условие:
Создай словарь, где ключ — первая буква слова, а значение — список всех слов,
начинающихся с этой буквы.
Пример:
words = ['apple', 'ant', 'banana', 'bee', 'cat']
Результат:
{
    'a': ['apple', 'ant'],
    'b': ['banana', 'bee'],
    'c': ['cat']
}
"""
def group_by_first_letter(words):
    result = {}
    for word in words:
        first_letter = word[0]
        if first_letter not in result:
            result[first_letter] = []
        result[first_letter].append(word)
    return result

words = ['apple', 'ant', 'banana', 'bee', 'cat']
print(group_by_first_letter(words))
```
