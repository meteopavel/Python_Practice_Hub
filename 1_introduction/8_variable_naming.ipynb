```python
#############################
### Именование переменных ###
#############################

"""
В именах переменных используйте только латинский алфавит, цифры и
подчеркивание (цифры нельзя ставить в начале имени переменной).
Остальные символы, в том числе буквы кириллицы, могут привести к ошибкам.

Если название состоит из нескольких слов, разделяйте их символом
подчёркивания: new_message. Такой стиль написания называется snake_case,
потому что слова могут получаться очень_длинные_как_змея_или_даже_как_две.

Альтернативный стиль именования называется camelCase. То есть каждое
последующее слово начинается с прописной буквы. Такой стиль принят
во многих языках (Java, JavaScript, TypeScrtipt, PHP). Но в Python
snake_case. Нужно придерживаться этого стиля, это стандарт.
"""
```

```python
"""
Python допускает использование цифр в именах переменных, но не на
первой позиции:
"""
snake1 = 'питон'
snake2 = 'удав'
long_snake = 'длинный змей'
print(long_snake + ' - это может быть ' + snake1 + ' или ' + snake2)
```

```python
"""
Названия переменных следует составлять из английских слов. Когда
программа вырастет до многих сотен строк, названия-слова помогут 
быстро сориентироваться в коде.

Например, можно дать имя переменной l, а можно lesson. Второй вариант 
намного лучше. Сокращение слова lesson до буквы l со временем забудется
и самим разработчиком, и тем более его коллегам.

Не рекомендуется использовать русские слова в английской раскладке. 
Рано или поздно код может быть прочитан человеком, не владеющим 
русским языком. Сразу нужно называть переменные по-английски: child 
вместо rebyonok.
"""
```
