```python
##############
### Списки ###
##############

"""
*Списки* — это последовательности чисел, строк или каких-то ещё значений.

Подобно тому, как мы присваивали переменным числовые и строковые значения,
точно так же переменная может содержать список.

Содержимое списка пишется в квадратных скобках, элементы списка 
разделяются запятой:

new_list = [<элемент>, <элемент>, <элемент>]
"""
# В переменной russian_alphabet сохраним список, состоящий из букв 
# алфавита. Буква - это строка, поэтому каждый элемент в кавычках
russian_alphabet = [
    'а','б','в','г','д','е','ё','ж','з','и','й','к','л','м','н','о','п',
    'р','с','т','у','ф','х','ц','ч','ш','щ','ъ','ы','ь','э','ю','я'
]
# Узнаем, что содержит переменная russian_alphabet:
print(russian_alphabet)
```

```python
"""
У каждого элемента есть свой порядковый номер - *индекс*. С помощью
индекса можно получить значение элемента списка: 
"""
russian_alphabet = [
    'а','б','в','г','д','е','ё','ж','з','и','й','к','л','м','н','о','п',
    'р','с','т','у','ф','х','ц','ч','ш','щ','ъ','ы','ь','э','ю','я'
]
print(russian_alphabet[1]) # напечатать элемент с индексом 1
print(russian_alphabet[2]) # напечатать элемент с индексом 2
"""
Неожиданно напечаталось б и в. Однако здесь нет ошибки. Индексация
в списках начинается с 0. То есть, если нужно напечатать букву а,
то следует написать:  
"""
print(russian_alphabet[0]) # напечатать элемент с индексом 0
```

```python
"""
Список из чисел может выглядеть так:
"""
countdown = [5, 4, 3, 2, 1, 0] 
print(countdown)
"""
Можно сделать список из выражений, тогда в нём будут храниться
вычисленные значения:
"""
# сохраним в списках вторую и третью строки таблицы Пифагора
pythagoras_2 = [ 
    2*1, 2*2, 2*3, 2*4, 2*5, 2*6, 2*7, 2*8, 2*9 
]
pythagoras_3 = [ 
    3*1, 3*2, 3*3, 3*4, 3*5, 3*6, 3*7, 3*8, 3*9 
]
print(pythagoras_2)
print(pythagoras_3)
```

```python
"""
К списку, который хранится в переменной, можно прибавить другой список.
Для примера наберём участников в группу "Бременские музыканты": 
"""
trubadur = ['Трубадур'] # Список может быть и из одного элемента;
animals = ['Кот', 'Пёс', 'Осёл', 'Петух']

bremen_musicians = trubadur + animals

print(bremen_musicians)
```

```python
"""
Для подсчёта элементов списка (и не только списка) есть стандартная
функция len():
"""
bremen_musicians = ['Трубадур', 'Кот', 'Пёс', 'Осёл', 'Петух']
count = len(bremen_musicians)
print(count)

my_str = 'somestr'
count = len(my_str)
print(count)
```

```python
"""
Задача 1:
Научите Анфису работать со списком друзей.
Для этого создайте список friends, содержащий элементы в таком порядке:
'Сергей', 'Соня', 'Дима', 'Алина', 'Егор'.
После того, как создадите список friends, напечатайте его содержимое.
"""
print('Привет, я Анфиса!')
friends = ['Сергей', 'Соня', 'Дима', 'Алина', 'Егор']
print(friends)
```

```python
"""
Задача 2:
Пора Анфисе стать вежливой и научиться здороваться. Анфиса, для начала
поздоровайся с Алиной!
Допишите код так, чтобы ваша программа напечатала фразу
Привет, Алина, я Анфиса!
"""
friends = ['Сергей', 'Соня', 'Дима', 'Алина', 'Егор']
index = 3
print('Привет, ' + friends[index] + ', я Анфиса!')
```

```python
"""
Задача 3:
Анфиса должна не только знать ваших друзей, но и рассказывать о каждом из них.
В переменной index записан номер друга, информация о котором нас интересует.
Получите из списка friends значение элемента с индексом, сохранённым в
переменной index и научите Анфису печатать сообщение: 
{имя друга с номером index} живёт в Красноярске
"""
print('Привет, я Анфиса!')
friends = ['Сергей', 'Соня', 'Дима', 'Алина', 'Егор']
index = 2
print(friends[index] + ' живёт в Красноярске.')
```

```python
"""
Задача 4:
Добавим Анфисе немного функциональности. Научим её считать друзей.
1. Объявите переменную count и сохраните в ней количество друзей.
Посчитайте их вызовом функции len().
2. Выведите на экран строку У тебя {количество} друзей, где
{количество} — значение переменной count.
Вывести эту фразу тремя способами: конкатенация, передача аргументов
через запятую, f-строка
"""
print('Привет, я Анфиса!')
friends = ['Сергей', 'Соня', 'Дима', 'Алина', 'Егор']
count = len(friends)
print('У тебя ' + str(count) + ' друзей')
print('У тебя', count, 'друзей')
print(f'У тебя {count} друзей')
```
