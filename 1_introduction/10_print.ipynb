```python
######################
### Вывод на экран ###
######################

"""
Мы уже умеем выводить на экран текст, складывая строки (конкатенация).
Но можно напечатать и без оператора +. Достаточно перечислить аргументы
функции print() через запятую:
"""
weather = 'облачно'
print('На улице сейчас', weather)
```

```python
"""
Запятая между аргументами по умолчанию заменяется на пробел. Однако
это не всегда удобно. Изучим поведение специального аргумента функции
print(). Это аргумент sep (от слова separate-'разделить')
"""
print('За', 'окном', 'метель') # эти строки напечатаются через пробел
gender = 'f'
if gender == 'm':
    str_end = ''
else:
    str_end = 'а'
print('Я увилел', str_end, ' за окном метель', sep='')
# именованные аргументы функции оформляются без пробелов вокруг =
print(f'Я увидел{str_end} за окном метель') # это пример f-строки

"""
Введя sep, мы изменяем символ между аргументами, так как в данном 
случае пробел по умолчанию мешал бы вывести строку корректно.
Необходимость указывать sep встречается редко. В реальных проектах
обычно используются средства форматирования строки.
"""
```

```python
"""
Следующий код демонстрирует то, что при передаче аргументов через
запятую нет необходимости приводить типы данных к строкам, как это
приходится делать в случае конкатенации:
"""
current_hour = 14
messages_count = 12
print('К', current_hour, 'ч.', 'у тебя накопилось',
      messages_count, 'новых сообщений.')
```
