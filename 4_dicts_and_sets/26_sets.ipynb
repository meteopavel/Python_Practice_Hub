```python
#################
### Множества ###
#################

"""
Ранее мы изучили списки:
"""
# Объявляем список bremen_musicians 
bremen_musicians = ['Трубадур', 'Кот', 'Пёс', 'Осёл', 'Петух']
"""
Помимо списков в Python есть и другие типы данных, которые могут 
хранить набор элементов. Такие типы данных называют *коллекции*.
"""
```

```python
"""
Одна из разновидностей коллекций в Python - это *множества*, тип
данных set. По структуре множества очень похожи на списки: в коде
они записываются как перечень элементов, разделённых запятой. Элементы
множества замыкаются в фигурные скобки {}.

Одно из важных отличий множества от других коллекций - все элементы 
множества должны быть уникальны, в множестве не может быть двух 
одинаковых элементов. 

Сравним, как ведут себя списки и сеты:
"""
concert_songs_list = [
    'Ничего на свете лучше нету',
    'Мы к вам заехали на час',
    'Рок-колыбельная',
    'Луч Солнца Золотого',
    'Ничего на свете лучше нету',
    'Куда ты, тропинка, меня завела',
    'А как известно, мы народ горячий'
]

concert_songs_set = {
    'Ничего на свете лучше нету',
    'Мы к вам заехали на час',
    'Рок-колыбельная',
    'Луч Солнца Золотого',
    'Ничего на свете лучше нету',
    'Куда ты, тропинка, меня завела',
    'А как известно, мы народ горячий'
}

print(concert_songs_list)
print(concert_songs_set)
```

```python
"""
Песня 'Ничего на свете лучше нету' в исходном коде упомянута дважды, но
на печать в случае сета она выводится лишь один раз: Python автоматически
удаляет из множества дубликаты, оставляя лишь уникальные значения.

Более того, изменяется порядов вывода элементов. При каждом запуске
порядок отличается от предыдущего. Это ещё одна особенность сетов: 
в памяти компьютера элементы сета хранятся в неупорядоченном виде и 
выводятся в случайном порядке.

Из этой особенности следует практический вывод: к элементу сета нельзя
обратиться по индексу, по порядковому номеру - один и тот же индекс
будет указывать на разные элементы (то на один, то на другой).

Проверим, что будет, если попробовать обратиться к элементу сета по
индексу:
"""
concert_songs_set = {
    'Ничего на свете лучше нету',
    'Мы к вам заехали на час',
    'Рок-колыбельная',
    'Луч Солнца Золотого',
    'Ничего на свете лучше нету',
    'Куда ты, тропинка, меня завела',
    'А как известно, мы народ горячий'
}

print(concert_songs_set[3])
```

```python
"""
Одно из полезных применений сетов заключается в избавлении дубликатов
в итерируемых объектах. Например, в списках. Предположим, что у нас есть 
список с песнями. Некоторые из повторяются. Нужно удалить дубликаты.
Это можно сделать с помощью функции set(), которая преобразует
итерируемый объект в множество:
"""
songs_list = [
    'Мы к вам заехали на час',
    'А как известно, мы народ горячий',
    'Куда ты, тропинка, меня завела',
    'Луч Солнца Золотого',
    'Рок-колыбельная',
    'Рок-колыбельная',
    'Куда ты, тропинка, меня завела',
    'А как известно, мы народ горячий',
    'Луч Солнца Золотого',
    'Ничего на свете лучше нету',
    'А как известно, мы народ горячий',
    'Луч Солнца Золотого',
    'Мы к вам заехали на час',
    'Ничего на свете лучше нету',
    'Куда ты, тропинка, меня завела',
    'Луч Солнца Золотого'
]

# Преобразуем список songs_list в сет
# и запишем этот сет в переменную unique_songs:
unique_songs = set(songs_list)

print(len(songs_list))
print(len(unique_songs))
print(unique_songs)
```

```python
"""
Множество содержит набор элементов, но получить к ним доступ по
индексу невозможно. Но какой смысл хранить данные без возможности
их извлечь и обработать?

Разумеется, выход есть: любую коллекцию можно перебрать в цикле,
а множество - это коллекция.

Объявление цикла и работа с ним для списков и множеств ничем не 
отличается. Но порядок, в котором будут обработаны элементы множества,
непредсказуем, так как множество - *неупорядоченная коллекция*.

Напечатаем названия песен в случайном порядке:
"""
unique_songs = {
    'А как известно, мы народ горячий',
    'Рок-колыбельная',
    'Ничего на свете лучше нету',
    'Куда ты, тропинка, меня завела',
    'Мы к вам заехали на час',
    'Луч Солнца Золотого'
}
# Сначала напечаем заголовок (пусть это будет афиша)
print('Концерт группы БРЕМЕНСКИЕ МУЗЫКАНТЫ!!!\n')

# Объявим цикл
for song in unique_songs:
    print(song)

print('\nНачало концерта в 19:00')
```
