```python
#################################
### Перебор элементов словаря ###
#################################

"""
Мы уже знаем как получить отдельный элемент словаря - с помощью доступа
по ключу. Однако если нужно получить все ключи или все значения, такой
метод уже не годится.

Возьмём словарь, созданный на предыдущем уроке. Затем получим два любых
его значения:
"""
friends = {
    'Евгений': 'Москва',
    'Михаил': 'Москва',
    'Иван': 'Казань',
    'Сергей': 'Новосибирск',
    'Тимофей': 'Санкт-Петербург',
    'Лиза': 'Санкт-Петербург',
    'Диана': 'Екатеринбург',
    'Ярослав':'Волгоград',
    'Алексей': 'Краснодар'
}

print(friends['Алексей'])
print(friends['Лиза'])
```

```python
"""
Словари, как и другие коллекции, можно перебрать в цикле  for...in

Словари в цикле обрабатываются по тому же принципу, что и списки,
но каждый элемент словаря состоит из двух частей, и поэтому при
объявлении цикла следует указать, что надо перебрать: только ключи,
только значения, или и то, и другое.

Если применить метод items(), то извлекаться будет и то и другое.
В таком случае следует объявить цикл сразу для двух переменных:
"""
friends = {
    'Евгений': 'Москва',
    'Михаил': 'Москва',
    'Иван': 'Казань',
    'Сергей': 'Новосибирск',
    'Тимофей': 'Санкт-Петербург',
    'Лиза': 'Санкт-Петербург',
    'Диана': 'Екатеринбург',
    'Ярослав':'Волгоград',
    'Алексей': 'Краснодар'
}

for name, city in friends.items():
    print(f'Мой друг {name} живёт в городе {city}')
```

```python
"""
При обработке словаря не обязательно извлекать и ключ, и значение 
каждого элемента. Можно перебрать только ключи словаря или только 
его значения. Для это есть методы keys() и values().

Если в цикле применём один из этих методов, тогда нужно объявлять 
только одну переменную цикла:
"""
friends = {
    'Евгений': 'Москва',
    'Михаил': 'Москва',
    'Иван': 'Казань',
    'Сергей': 'Новосибирск',
    'Тимофей': 'Санкт-Петербург',
    'Лиза': 'Санкт-Петербург',
    'Диана': 'Екатеринбург',
    'Ярослав':'Волгоград',
    'Алексей': 'Краснодар'
}

print('У меня есть друзья:')
for name in friends.keys():
    print(name)

print('\nОни живут в городах:')
for city in friends.values():
    print(city)

```

```python
"""
По умолчанию Python для итерации по словарю использует ключи. Поэтому
прописывать метод keys() не обязательно
"""
friends = {
    'Евгений': 'Москва',
    'Михаил': 'Москва',
    'Иван': 'Казань',
    'Сергей': 'Новосибирск',
    'Тимофей': 'Санкт-Петербург',
    'Лиза': 'Санкт-Петербург',
    'Диана': 'Екатеринбург',
    'Ярослав':'Волгоград',
    'Алексей': 'Краснодар'
}

print('У меня есть друзья:')
for name in friends:
    print(name)
```

```python
"""
Задача 1.
Сделать два списка. Один - имена друзей. Второй - города, где они живут.
Объявить пустой словарь friends и заполнить его элементами списка.
Имена должны стать ключами, а города - значениями.
Напечатать сообщение: Сергей живёт в городе Новосибирск. При этом 
использовать доступ по нужному ключу в словаре friends
"""
friends_name = ['Евгений', 'Михаил', 'Иван', 'Сергей', 'Тимофей',
                'Лиза', 'Диана', 'Ярослав', 'Алексей']
friends_city = ['Москва', 'Москва', 'Казань', 'Новосибирск',
                'Санкт-Петербург', 'Санкт-Петербург', 'Екатеринбург',
                'Волгоград', 'Краснодар']

friends = {}

for i in range(len(friends_name)):
    friends[friends_name[i]] = friends_city[i]

print('Сергей живёт в городе {}'.format(friends['Сергей']))
```

```python
"""
Задача 2.
Напечатать о каждом из друзей такое сообщение:
<имя_друга> живёт в городе <название_города>
Нужно использовать метод items().  
"""
friends = {
    'Евгений': 'Москва',
    'Михаил': 'Москва',
    'Иван': 'Казань',
    'Сергей': 'Новосибирск',
    'Тимофей': 'Санкт-Петербург',
    'Лиза': 'Санкт-Петербург',
    'Диана': 'Екатеринбург',
    'Ярослав':'Волгоград',
    'Алексей': 'Краснодар'
}

for name, city in friends.items():
    print(f'{name} живёт в городе {city}')
```

```python
"""
Задача 3.
Подготовить словарь, в котором ключами будут имена людей (3 человека), 
а значениями - их любимые песни (у каждого 2-4 песни).

Затем нужно будет напечатать количество любимых песен одного человека
и после этой строки напечатать все любимые песни другого человека
построчно.
"""
friends = {
    'Коля': ['Ринопластика', 'HELL'],
    'Матвей': ['Phoenix', 'So Good'],
    'Влад': ['2017', 'LMFAO']
}

person = 'Коля'
print(f'Количество любимых песен {person}: {len(friends[person])}')

other_person = 'Матвей'
print(f'Любимые песни {other_person}:')
for song in friends[other_person]:
    print(song)
```
