```python
##############################
### Операции с множествами ###
##############################

"""
*Метод* - это разновидность функции, мини-программа. В отличие от 
функций, метод применяется к *объекту*, например - к строке, к числу
или к множеству. У строк - свои методы, у чисел - свои,
у множеств - свои.
"""
```

```python
"""
Чтобы добавить элемент в множество, к нему применяют метод add(). В
качестве аргумента он принимает значение, которое нужно добавить. Добавить
можно только один элемент, иначе придётся вызывать метод несколько раз
с разными аргументами. 
Применим метод add() к объекту playlist. Аргументом будет строка
'Thunderstruck' 
"""
playlist = {
    'Venus',
    'Yesterday',
    'Fireball',
    'Time',
    'Poison'
}

playlist.add('Thunderstruck')
print(playlist)
```

```python
"""
Множества в Python можно объединять. Например, можно составить из двух 
плейлистов один общий, чтобы ни одна песня не повторялась.

Для этого применяют метод union(). Он применяется к одному множеству,
а в аргументе передаётся второе.

В результате будет создано новое, третье множество (а оба исходных
останутся без изменений).
"""
playlist_1 = {'Три белых коня', 'Happy new year', 'Снежинка'}
playlist_2 = {'Last christmas', 'Снежинка', 'Happy new year'}
playlist_3 = playlist_1.union(playlist_2)

print(playlist_3)
```

```python
"""
Если нужно найти песни, которые присутствуют в одном плейлисте, но
отсутствуют во втором - поможет метод difference().

Метод set_1.difference(set_2) вернёт новое множество, оно будет содержать
только те элементы, которые присутсвуют в set_1, но отсутствуют в set_2.
Это похоже на вычитание (set_1 - set_2).

Исходные множества не изменятся.
"""
playlist_1 = {'Голубой вагон', 'Облака', 'Yesteday', 'Наше лето'}
playlist_2 = {'Наше лето', 'Голубой вагон', 'Облака'}
playlist_3 = playlist_1.difference(playlist_2)

print(playlist_3)
```

```python
"""
Можно найти пересечение двух множеств, то есть элементы, которые есть и 
в первом и во втором множестве.

Посмотрим на список фильмов, просмотренных двумя людьми и определим, в 
чём их вкусы совпадают. Для этого применим метод intersection(), он вернёт
новое множество, в котором будут храниться пересекающиеся элементы.
"""
films_1 = {'Форсаж', 'Достучаться до небес', 'Мстители'}
films_2 = {'Мстители', 'Форсаж', 'Матрица'}
films_3 = films_1.intersection(films_2)

print(films_3)
```

```python
"""
Задача 1.
Научим Анфису играть в города. У неё в памяти хранится два перечня городов:
- в множестве all_cities хранятся все города, которые она знает;
- в множестве used_cities - города, которые уже были названы, их использовать
уже нельзя.

Для начала научим Анфису получать перечень городов, которые ещё не были
названы в игре. 

Объявим функцию print_valid_cities(). Она должна принимать два аргумента - 
множества all_cities и used_cities. Затем функция должна создать третье
множество - разница между входными. А затем функция должна построчно 
напечатать элементы результирующего множества.
"""
all_cities = {
    'Абакан',
    'Астрахань', 
    'Бобруйск', 
    'Калуга',
    'Караганда',
    'Кострома',
    'Липецк', 
    'Новосибирск'
}

used_cities = {'Калуга', 'Абакан' , 'Новосибирск'}

def print_valid_cities(all_cities, used_cities):
    valid_cities = all_cities.difference(used_cities)
    print(valid_cities)
    

print_valid_cities(all_cities, used_cities)
```

```python
"""
Задача 2.
Продолжаем играть с Анфисой в города. Теперь нужно добавить дополнительный
список городов для игры. 

Необходимо написать новую функцию add_cities(). Она добавит элементы из
списка new_cities в all_cities.

Метод union() для этой задачи не подойдёт, поскольку нам нужно добавить 
элементы в существующее множество, а не создать новое. 
"""
new_cities = [
    'Екатеринбург',
    'Выборг' ,
    'Владивосток',
    'Казань',
    'Why',
    'Йезд'
]

all_cities = {
    'Абакан',
    'Астрахань', 
    'Бобруйск', 
    'Калуга',
    'Караганда',
    'Кострома',
    'Липецк', 
    'Новосибирск'
}

used_cities = {'Калуга', 'Абакан' , 'Новосибирск'}

def print_valid_cities(all_cities, used_cities):
    valid_cities = all_cities.difference(used_cities)
    for city in valid_cities:
        print(city)


def add_cities(all_cities, new_cities):
    for city in new_cities:
        all_cities.add(city)

add_cities(all_cities, new_cities)
print_valid_cities(all_cities, used_cities)
```

```python
"""
Задача 3.
Бот Анфиса и бот Алиса хотят сыграть во что-нибудь по сети. Каждая из
них составила список игр, в которые она умеет играть. Списки разные, но,
возможно, найдутся игры, которые знакомы им обеим.

Нужно написать программу, которая найдёт одинаковые элементы в двух
списках.

Допишите функцию get_toghether_games(): она должна принимать на вход 
два списка, а возвращать - множество игр, названия которых есть в обоих
списках.

Функция должна напечатать эти игры построчно, и перед каждой из них
поставить эмоджи 👾 и пробел. 

Результат должен выглядеть примерно так:
👾 Super Hero Developer
👾 Python Shooter
👾 Online-backgammon 
"""
anfisa_games = [
    'Online-chess',
    'Города',
    'DOOM',
    'Крестики-нолики'
]
alisa_games = [
    'DOOM',
    'Online-chess',
    'Города',
    'GTA',
    'World of tanks'
]

def get_together_games(anfisa_games, alisa_games):
    anfisa_games = set(anfisa_games)
    alisa_games = set(alisa_games)
    together_games = anfisa_games.intersection(alisa_games)
    return together_games

together_games = get_together_games(anfisa_games, alisa_games)

for game in together_games:
    print(f'👾 {game}')
```
