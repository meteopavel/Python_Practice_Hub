```python
######################################
### Составные логические выражения ###
######################################

"""
Логические операторы можно объединять в составные выражения. В таких
выражениях операторы выполняются не в порядке записи, а в порядке 
приоритета: высший приоритет у not, затем выполняется and, а последним or

Оператор not - это логическое отрицание, он возвращает True, если выражение
равно False и наоборот. Для примера можно вывести на печать результаты
применения оператора not:  
"""
print(True)
print(not True) # False
print(not False) # True
```

```python
"""
Любая операция сравнения возвращает True или False. А оператор not просто
превращает эти значения в их противоположности:
"""
print(5 > 3) # True
print(not 5 > 3) # False
```

```python
"""
Протестируем not на практике. Напечатаем стихотворение о погоде. Текст 
будет зависеть от переменной wind:
"""
wind = True

if not wind: # Если wind НЕ равен True
    print('Ночь тиха')
else:
    print('Поднялся ветер')
    print('Серые тучи развеял')
    print('Новые тянутся с юга')
```

```python
"""
В любом составном выражении сначала выполняется оператор not, затем
and, а последним or.

Рассмотрим следующий код и выясним, что будет напечатано:
"""
x = 44

if not x > 40 or x < 45 and x != 42:
    print('Выражение вернуло True')
else:
    print('Выражение вернуло False')

"""
* Сперва выполнится not, значит, вместо выражения not x > 40 будет False
(x = 44, 44 > 40 == True), а not True == False);
* Потом выполнится оператор and: выражение 44 < 45 and 44 != 42 вернет True;
* В конце сработает or: False or True вернёт True. 
"""
```

```python
"""
Напишем программу, которая по совокупности данных определяет погоду
и выводит на экран сообщение "Идём гулять, на улице хорошо", если:
* не идёт дождь ИЛИ не сильно ветренно (главное, чтобы не было ветра
одновременно с дождём),
* температура воздуха более 22 градусов.
Программа будет состоять из нескольких логических выражений:
"""
# Передадим в программу данные
beaufort = 6         # Сильный ветер
is_raining = False   # Нет дождя
temperature = 16     # Не жарко

if not is_raining or beaufort <= 4 and temperature > 22:
    print('Идём гулять, на улице хорошо')
else:
    print('Сидим дома, учим Питон')
# Результат неверный, что-то пошло не так
```

```python
"""
В предыдущем коде вначале вычисляется значение каждого выражения,
в том числе и с not:
not is_raining    - результат True
beaufort <= 4     - резульат False
temperature > 22  - резульат False

Затем вычисляются оставшиеся логические операторы, сначала and, так как
у него приоритет выше:
beaufort <= 4 and temperature > 22 -> False and False -> False

Далее идёт проверка: если истинно * хотя бы одно * из условий,
not is_raining или beaufort <= 4 and temperature > 22 - сообщается
о хорошей погоде.

Выражение not is_raining истинно, поэтому выводится сообщение о 
хорошей погоде, хотя на самом деле всё наоборот.

Чтобы исправить код, необходимо заключить в скобки первые два сравнения:
"""
# Передадим в программу данные
beaufort = 6         # Сильный ветер
is_raining = False   # Нет дождя
temperature = 16     # Не жарко

if (not is_raining or beaufort <= 4) and temperature > 22:
    print('Идём гулять, на улице хорошо')
else:
    print('Сидим дома, учим Питон')
```
