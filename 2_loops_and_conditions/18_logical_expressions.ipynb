```python
############################
### Логические выражения ###
############################

"""
Кроме простых операторов сравнения
* "равно" ==,
* "меньше" <,
* "больше" >,
есть и более сложные операторы, учитывающие сразу два условия:
* "больше или равно" >=,
* "меньше или равно" <=,
* "не равно" !=.
"""
beaufort = 4

if beaufort >= 6:
    print('Будьте осторожны, сильный ветер!')

if beaufort <= 4:
    print('Легкий ветер')

if beaufort != 0:
    print('Ветренно')
```

```python
"""
Когда нужно выбирать между двумя условиями, в ход идут *логические
операции*.

Оператор or, "Логическое ИЛИ", возвращает True, если *хотя бы одно*
из условий истинно
"""
beaufort = 7

# Если ветер в 7 или 8 баллов, то он называется крепким
if beaufort == 0:
    print('Штиль')
elif beaufort == 1:
    print('Тихий ветер')
elif beaufort == 2:
    print('Лёгкий ветер')
elif beaufort == 3:
    print('Слабый ветер')
elif beaufort == 4:
    print('Умеренный ветер')
elif beaufort == 5:
    print('Свежий ветер')
elif beaufort == 6:
    print('Сильный ветер')
elif beaufort == 7 or beaufort == 8:
    print('Крепкий ветер')

"""
Если первый аргумент оператора or возвращает True, то второй 
аргумент не оценивается, так как ответ уже ясен.
"""
```

```python
"""
Оператор and, "Логическое И", возвращает True, только если *оба*
условия истинны:
"""
beaufort = 9

# если ветер от 9 до 11 баллов, то это шторм (различной силы)
if beaufort == 0:
    print('Штиль')
elif beaufort == 1:
    print('Тихий ветер')
elif beaufort == 2:
    print('Лёгкий ветер')
elif beaufort == 3:
    print('Слабый ветер')
elif beaufort == 4:
    print('Умеренный ветер')
elif beaufort == 5:
    print('Свежий ветер')
elif beaufort == 6:
    print('Сильный')
elif beaufort == 7 or beaufort == 8:
    print('Крепкий ветер')
elif beaufort >= 9 and beaufort <= 11:
    print('Шторм')
elif beaufort == 12:
    print('Ураган')
else:
    print('Неизвестное значение')

"""
Если один из аргументов оператора and возвращает False, то другой
аргумент не оценивается, так как оператор в любом случае вернёт 
False
"""
```
