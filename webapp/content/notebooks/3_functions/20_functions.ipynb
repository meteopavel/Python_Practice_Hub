```python
###############
### Функции ###
###############

"""
Иногда в разных частях программы нужно выполнять одинаковые действия.
Самый простой подход в такой ситуации - взять готовый кусок кода и 
скопировать его туда, где требуется повторить выполнение.

Например, раз в минуту Анфиса должна проверять входящую почту: если
появились новые письма - переменная new_mail станет больше нуля, и 
Анфиса сообщит об этом.
"""
new_mail = 0
if new_mail > 0:
    print('Пришло письмо, не пропусти!')
else:
    print('Никто не пишет.')
# Тут код, отсчитывающий минуту
if new_mail > 0:
    print('Пришло письмо, не пропусти!')
else:
    print('Никто не пишет.')
# Тут код, отсчитывающий минуту
if new_mail > 0:
    print('Пришло письмо, не пропусти!')
else:
    print('Никто не пишет.')
# Пусть за это время пришло письмо:
new_mail = 1
# И снова проверка
if new_mail > 0:
    print('Пришло письмо, не пропусти!')
else:
    print('Никто не пишет.')
```

```python
"""
Предыдущий код работает, но если потребуется проверка в течение суток?
Если программа будет достаточно сложная, то такое дублирование быстро
приведёт к тому, что код станет невозможно читать и поддерживать.

От таких проблем спасают * функции *.

* Функция * - это именованный блок кода, выполняющий определённую задачу.
Код функции можно использовать многократно, надо лишь вызвать её - обра-
титься к ней по имени. 

В Python есть множество втроенных функци, некоторые мы уже вызывали:
print(), str(), int(), float(), len().

Но можно создавать и собственные функции.
"""
new_mail = 0
# Здесь мы объявляем функцию и описываем что она делает
def check_mail():
    if new_mail > 0:
        print('Пришло письмо, не пропусти!')
    else:
        print('Никто не пишет.')
# Здесь мы вызываем функцию
check_mail()
# Тут код, отсчитывающий минуту
check_mail()
# Тут код, отсчитывающий минуту
check_mail()
# Поступило новое письмо
new_mail = 1
# Снова вызываем функцию
check_mail()
```

```python
"""
Предыдущий код работает точно также, но при этом намного лаконичнее и
понятнее.

Функцию объявляют ключевым словом def, а затем указывают * имя * функции
(его придумывает разработчик), после имени - круглые скобки и двоеточие.

За объявлением функции следует код, который функция ддолжна выполнить при 
вызове. Этот код называет * телом функции *. Тело функции, как и тело цикла,
отбивается четырьмя пробелами: по этим отступам Python понимает, где 
начинается и заканчивается тело функции. 

Мало просто объявить функцию. Если выполнить код, в котором только её 
объявление - ничего не произойдёт. Функция не выполняется до тех пор,
пока её не вызвать. Она будет просто лежать и ждать своего часа.

Чтобы вызвать функцию - нужно "позвать" её по имени: написать название 
функции и после названия поставить круглые скобки.

Каждый вызов функции должен быть на отдельной строке.
 """
def hello():
    print('Приветствую тебя, Джедай Питона!')
# Пока следующие строки закоментированны, при запуске программы ничего не 
# произойдёт. 
hello()
hello()
hello()
```

```python
"""
Пример выше очень простой: при вызове функция hello() постоянно выводит 
одну и ту же строку.

А что делать, если нужно поздороваться, но в приветсвии упомянуть имя:
"Максим, приветствую тебя!" ?

С одной сторны, можно так:
"""
def hello():
    print('Максим, приветствую тебя!')

hello()
"""
Но тогда придётся для каждого имени писать новую функцию. Хорошо бы вместо
имени подставить переменную:
"""
def hello():
    print(name + ', приветствую тебя!')

hello()
"""
Это красиво, но код не сработает: функция не знает, что за переменная name
и какое у неё значение. 
"""
```

```python
"""
Решение есть: при объявлении функции можно указывать * параметры функции *.
Это такие переменные, которые будут обрабатываться в её теле. Имена для 
параметров придумывает разработчик.

Значения для этих параметров передаются при вызове - их называют 
* аргументами функции *. 

Параметры указывают в круглых скобках при объявлении функции; аргументы 
указывают при вызове функции.

В следующем коде при вызове функции строка "Максим" будет присвоена
переменной name, это значение будет подставлено в строку, которую напечатает
функция print():
"""
# Объявляем функцию с параметром name
def hello(name):
    # Параметр name можно обрабатывать в теле функции
    print(name + ', приветствую тебя!')
# Вызов функции с аргументом 'Максим'
hello('Максим')
# Будет напечатано: Максим, приветствую тебя!
# А можно поприветствовать и себя самого
hello('Павел')
```

```python
"""
У функции может быть несколько параметров, в этом случае они перечисляются
через запятую. При вызове функции аргументы передаются в соответствии с
порядком записи (тоже через запятую): первый аргумент передаётся в первый
параметр, второй аргумент - во второй параметр.
Сделаем такую функцию:
"""
# Теперь у функции hello() два параметра: name и bonus
def hello(name, bonus):
    # Оба параметра применим в теле функции
    print(name + ', приветствую тебя! Бери ' + bonus)

# Создадим словарь (доп. материал)
names = {
    'Дарт Вейдер': 'Звезду Смерти', 
    'Винну Пух': 'Мёд', 
    'Павел': 'Печеньки',
    'Саша': 'Йогурт'
}

# Переберем в цикле словарь и для каждой пары ключ-значение вызовем hello()
for key in names:
    hello(key, names[key])
```

```python
"""
Функция - это такой же составной блок программы, как циклы и операторы
ветвления. Их можно комбинировать между собой. Например, в коде предыдущей
программы мы вызывали функцию внутри тела цикла.

А можно наоборот: переписать программу так, чтобы цикл был внутри тела 
функции:
"""
def hello(dictionary):
    for key in dictionary:
        print(key + ', приветствую тебя! Бери ' + dictionary[key])

names = {
    'Дарт Вейдер': 'Звезду Смерти', 
    'Винну Пух': 'Мёд', 
    'Павел': 'Печеньки',
    'Саша': 'Йогурт'
}

hello(names)

"""
Искусство программирования заключается в том, как разработчик комбинирует
разные блоки кода друг с другом.
"""
```

```python
"""
Задача 1
Напишите функцию print_friends_count() для вывода количества друзей.
При вызове в функцию должно передаваться количество друзей.
Параметр функции должен называться friends_count.

Вызовите эту функцию с разными аргументами не менее трёх раз.
Функция должна сообщать о количестве друзей при любых целых
положительных значениях переменной friends_count.

Для количества друзей < 20, фраза должна корректно склоняться.
Если же друзей очень много — двадцать или больше, — должно
выводиться сообщение:
'Ого, сколько у тебя друзей! Целых {friends_count}'

Например:
У тебя нет друзей
У тебя 1 друг
У тебя 2 друга
У тебя 6 друзей
Ого, сколько у тебя друзей! Целых 20 
"""
def print_friends_count(friends_count):
    if friends_count == 0: 
        print('У тебя нет друзей') 
    elif friends_count == 1: 
        print('У тебя', friends_count, 'друг') 
    elif friends_count >= 2 and friends_count <= 4: 
        print('У тебя', friends_count, 'друга') 
    elif friends_count >= 5 and friends_count < 20: 
        print('У тебя', friends_count, 'друзей') 
    else: 
        print('Ого, сколько у тебя друзей! Целых', friends_count)


print_friends_count(0)
print_friends_count(1)
print_friends_count(2)
print_friends_count(6)
print_friends_count(20)
```
