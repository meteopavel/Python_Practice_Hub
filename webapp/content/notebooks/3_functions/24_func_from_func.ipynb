```python
################################
### Вызов функции из функции ###
################################

"""
На предыдущем этапе мы сделаи программу для расчёта метража рёбер 
и площади стекла для кубов. Программа работает, но в ней есть недочёты:
работать с кодом неудобно, фрагменты кода повторяются. 

Приходится отдельно вызывать функции для расчета метража и площади. При
этом, вызывая функции, в них передаётся один и тот же аргумент - длина
ребра куба (3).

Афанасию лучше будет написать функцию calc_cube(), которая сама будет
вызывать обе эти функции и печатать калькуляцию необходимых материалов.
Тогда можно будет обойтись лишь одним вызовом:
"""
# Функция для расчёта периметра сторон куба
def calc_cube_perimeter(side):
    return side * 12

# Функция для расчёта площади остекления граней куба
def calc_cube_area(side):
    return side * side * 6

# Основная функция, которая принимает длину ребра куба
def calc_cube(side):
    # Вызываем функцию, рассчитывающую периметр
    # и передаём в неё длину стороны куба
    one_cube_perimeter = calc_cube_perimeter(side)
    full_length = one_cube_perimeter * 8

    # Вызываем функцию, рассчитывающую площадь остекления
    # и передаём в неё длину стороны куба
    one_cube_area = calc_cube_area(side)
    full_area = one_cube_area * 8

    print('Для 8 кубов понадобится рёбер (м):', full_length, 
          'и стекла (кв.м)', full_area) 
    
# В результате остался лишь один вызов "основной" функции,
# а она уже вызовет все вспомогательные
calc_cube(3)
```

```python
"""
Вызов функции из другой фукции- распространённый приём, это позволяет
сделать код более структурированным и универсальным. В программе
получается одна "точка входа" - стартовый код, в который передаются
все необходимые данные. В программе Афанасия такой точкой входа
служит функция calc_cube().

Есть ещё одно преимущество: размер куба задаётся лишь в одном месте
кода - при вызове calc_cube(). Это уменьшает вероятность ошибки и 
упрощает работу с программой.
"""
```

```python
"""
Задача 1. ДЗ
Необходимо изменить программу Афанасия так, чтобы можно было передавать
не только сторону куба, но и количество кубов. Сейчас программа делает 
расчёт только для 8 кубов, а если потребуется другое их количество - 
придётся менять код основной функции. 

Нужно добавить к функции calc_cube() новый параметр, назовём его amount -
количество кубов. Этот параметр используем в нужных местах кода функции,
в том числе и в print().

Так программа станет ещё более универсальной.
"""
def calc_cube_perimeter(side):
    return side * 12

def calc_cube_area(side):
    return side * side * 6

def calc_cube(side):
    one_cube_perimeter = calc_cube_perimeter(side)
    full_length = one_cube_perimeter * 8

    one_cube_area = calc_cube_area(side)
    full_area = one_cube_area * 8

    print('Для 8 кубов понадобится рёбер (м):', full_length, 
          'и стекла (кв.м)', full_area) 
    
calc_cube(3)
```
