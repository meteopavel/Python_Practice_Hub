```python
#############
### Циклы ###
#############

"""
В обычной жизни нам часто приходится выполнять серии одинаковых действий.
Например, помыть тарелки, почистить картофель, пожарить блины и т.д.
Эти действия можно декомпозировать на поддействия. Напрмер, каждую тарелку
необходимо намочить, натереть моющим средством, прополоскать, вытереть 
и поставить на место. Так для каждой тарелки.

Этот процесс не бесконечен: работа должна продолжаться при условии, что 
остались немытые тарелки. Если они закончились - работа прекращается.

В разработке тоже приходится выполнять одинаковые действия. Например, 
часто требуется произвести некоторые операции над каждым элементом 
итерируемого объекта. Забегая вперёд скажем, что ранее изученные списки
как раз являются итерируемыми. Иначе можно сказать перечисляемыми, состоящими
из набора элементов.
"""
```

```python
"""
Возьмём для примера список уже знакомых бременских музыкантов:

bremen_musicians = ['Кот', 'Пёс', 'Трубадур', 'Осёл', 'Петух']

Задача состоит в том, чтобы напечатать каждого на отдельной строке:
Кот
Пёс
Трубадур
Осёл
Петух
"""
```

```python
"""
В Python это можно сделать, прописывая вручную вывод каждого элемента,
обращаясь к нему по индексу:
"""
bremen_musicians = ['Кот', 'Пёс', 'Трубадур', 'Осёл', 'Петух']
print(bremen_musicians[0])
print(bremen_musicians[1])
print(bremen_musicians[2])
print(bremen_musicians[3])
print(bremen_musicians[4])
"""
Пришлось писать пять раз один и тот же код. А в случае более объёмного списка
пришлось бы делать это намного большее количество раз.

Кроме того, итерируемые объекты могут меняться в процессе исполнения 
программы. Поэтому такой подход весьма неэффективен.
"""
```

```python
"""
Для выполнения подобных однотипных операций, зависящих только от конкретных
параметров (в задаче выше этот параметр - индекс списка), придумали *циклы*.

*Циклы* - это программные конструкции, выполняющие определённые действия 
до тех пор, пока выполяется заданое условие.

Чтобы программа поняла, что сейчас начнётся цикл - нужно сообщить ей об этом
специальными словами: *объявить цикл*. В Python есть несколько разновидностей 
циклов, но перебирать список проще всего циклом for ... in ...

Покажем, как предыдущую задачу можно решить циклом (один из вариантов):
"""
bremen_musicians = ['Кот', 'Пёс', 'Трубадур', 'Осёл', 'Петух']
for musician in bremen_musicians:
    print(musician)
```

```python
"""
Цикл в Python объявляется ключевыми словами for и in. После объявления 
ставится двоеточие. Ниже объявления пишут *тело цикла* - код, который
описывает, что нужно сделать на каждом этапе цикла. Этап цикла называется
*итерацией*.

Сейчас мы работаем со списками. Поэтому будем рассматривать работу цикла
применительно к ним. Хотя на самом деле, это будет справедливо и для других
итерируемых объектов.

После слова for указывают имя переменной, в которую будут поочерёдно 
передаваться элементы из обрабатываемого списка. А после in ставится имя 
списка, который надо обработать.

Имя переменной можно указать любое. Но традиционно делают так: имена списков 
указывают во множественном числе, и имена переменных в единственном. Так 
получается хорошо читаемый код: для каждого музыканта в Бременских музыкантах
(for musician in bremen_musicians:)

Python самостоятельно берёт на себя работу со списком, и цикл завершит 
свою работу, когда список закончится.
"""
```

```python
"""
Тело цикла. 
После объявления цикла пишут его тело. Каждая строка тела обязательно
отбивается от начала строки четырьмя пробелами. Всё, что в теле цикла,
(переменные, объявленные внутри) недоступно за его пределами. В других 
языках принадлежность к телу конструкции обычно определяется скобками,
но в Python именно отступы определяют эту принадлежность.
"""
```

```python
"""
Рассмотри подробно уже знакомый код:
"""
# Объявляем и заполняем список с музыкантами. При добавлении новых
# участников, все они также попадут в печать
bremen_musicians = ['Кот', 'Пёс', 'Трубадур', 'Осёл', 'Петух']
# Код цикла
for musician in bremen_musicians:
    # Каждый элемент списка bremen_musicians
    # по очереди будет передан в переменную musician
    # и затем напечатан
    print(musician)

# Код, который будет выполнен уже после цикла
print('Нам дворцов заманчивые своды не заменят никогда свободы!')
```

```python
"""
Задача 1. Напечатать с помощью цикла названия месяцев из прилагаемого
списка
"""
months = ['Январь', 'Февраль', 'Март', 'Апрель', 'Май', 'Июнь', 'Июль', 'Август', 'Сентябрь', 'Октябрь', 'Ноябрь', 'Декабрь']
for month in months:
    print(month)
```

```python
"""
Задача 2. 
Напечатайте приглашение на ужин: циклом выведите имена приглашённых
из списка pigs
Текст должен быть следующий:
Дорогие свиньи!
Ниф-Ниф
Наф-Наф
Нуф-Нуф
приглашаю вас на ужин!
Любящий вас Волк.
"""
pigs = ['Ниф-Ниф', 'Наф-Наф', 'Нуф-Нуф']
print('Дорогие свиньи!')
for pig in pigs:
    print(pig)
print('приглашаю вас на ужин!')
print('Любящий вас Волк.')
```

```python
"""
Задача 3.
Напишите программу, которая запрашивает у пользователя целое число n
и затем вычисляет сумму всех целых чисел от 1 до n (включительно)
с помощью цикла for. 

Пример работы:
Введите целое число n: 5
Сумма чисел от 1 до 5 равна 15.

Подсказка:
Для вычисления суммы используйте переменную, в которую будете
накапливать сумму внутри цикла.

Понадобится функция range(). Функция range() в Python используется для
генерации последовательности чисел. 
range(start, stop, step)
Пример 1:
for i in range(5):
    print(i)
Вывод: 0, 1, 2, 3, 4
Пример 2:
for i in range(2, 6):
    print(i)
Вывод: 2, 3, 4, 5
"""
n = int(input('Введите число, получите сумму от нуля до этого числа: '))
sum = 0
for i in range(1, n + 1):
    sum += i
print(sum)
```

```python
"""
Задача 4.
Напишите программу, которая выводит таблицу квадратов чисел от 1 до 10.

Пример вывода программы:
1^2 = 1
2^2 = 4
...
10^2 = 100

Возведение в степень можно сделать оператором **
"""
power = 2
for n in range(1, 11) :
    result = n ** power
    print(str(n) + '^2 = ' + str(result))
```

```python
"""
Задача 5.
Напишите программу, которая вычисляет сумму квадратов первых n
натуральных чисел. Значение n вводится пользователем.

Пример выполнения:
Введите значение n: 5
Сумма квадратов первых 5 натуральных чисел равна 55
"""
n = int(input('Введите значение n: '))
sum_squres = 0
for i in range(1, n + 1):
    sum_squres += i ** 2
print('Сумма квадратов первых', n, 'натуральных чисел равна', sum_squres)
```
