```python
###############################
### Множественные ветвления ###
###############################

"""
Для нормальной оценки силы ветра по шкале Бофорта нужно учитывать
градации. Можно сделать такую программу с конструкцией else:
"""
beaufort = 0
if beaufort == 0:
    print('Штиль')
else:
    if beaufort == 1:
        print('Тихий ветер')
    else:
        if beaufort == 2:
            print('Лёгкий ветер')
        else:
            if beaufort == 3:
                print('Слабый ветер')
            else:
                if beaufort == 4:
                    print('Умеренный ветер')
                else:
                    if beaufort == 5:
                        print('Свежий ветер')
                    else:
                        if beaufort == 6:
                            print('Сильный ветер')
                        else:
                            print('Шторм')
```

```python
"""
Этот код состоит из множества вложенных друг в друга условных 
операторов. Каждый блок отбивается четырьмя пробелами, что усложняет
и без того непростое восприятие.

Подобный код необходимо упростить, применив конструкцию elif. Эта
конструкция является вложенной для if. Если не выполняется if - управление
переходит в elif.

Для elif необходимо прописывать своё логическое выражение:
"""
beaufort = 0
if beaufort == 0:
    print('Штиль')
elif beaufort == 1:
    print('Тихий ветер')
elif beaufort == 2:
    print('Лёгкий ветер')
elif beaufort == 3:
    print('Слабый ветер')
elif beaufort == 4:
    print('Умеренный ветер')     
elif beaufort == 5:
    print('Свежий ветер')
elif beaufort == 6:
    print('Сильный ветер')
else:
    print('Шторм')
"""
Нужно помнить про "ленивость" вычислений: как только выполняется 
одно из условий, все последующие elif и else игнорируются. Поэтому 
необходимо расставлять elif в порядке приоритетности для корректной
работы программы.
"""
```

```python
"""
Задача 1.
Нужно просклонять слово "сообщения" в зависимости от их количества:
для 0 - "У вас нет новых сообщений"
для 1 - "У вас 1 новое сообщение"
от 2 до 4 включительно - "У вас <количество> новых сообщения"
любое другое количество - "У вас <количество> новых сообщений"
Пусть будет до 20 сообщений.
"""
for messages_count in range(21): 
    if messages_count == 0: 
        print('У вас нет новых сообщений') 
    elif messages_count == 1: 
        print('У вас 1 новое сообщение') 
    elif messages_count <= 4: 
        print('У вас', messages_count, 'новых сообщения') 
    else: 
        print('У вас', messages_count, 'новых сообщений')
```

```python
"""
Задача 2.
Нужно научить Анфису желать доброго утра, дня, вечера и ночи. 
Понадобится соответствующий цикл. 
Если значение переменной (назовём её current_hour) 
строго меньше 6 - написать сообщение "Доброй ночи!". 
От 6 включительно, но строго меньше 12 - "Доброе утро!"
От 12 включительно, но строго меньше 18 - "Добрый день!"
От 18 включительно, но строго меньше 23 - "Добрый вечер!"
Перед пожеланием нужно вывести фразу: "На часах <время>"
Время представить в формате 12:00

Пример:
...
На часах 9:00. 
Доброе утро!
...
"""
for current_hour in range(24):
    print('На часах ' + str(current_hour) + ':00')
    if current_hour < 6:
        print('Доброй ночи!')
    elif current_hour < 12:
        print('Доброе утро!')
    elif current_hour < 18:
        print('Добрый день!')
    elif current_hour < 23:
        print('Добрый вечер!')
    else:
        print('Доброй ночи!')
```
