```python
#################
### Ветвления ###
#################

"""
Кроме арифметических выражений, в Python есть *логические*. Они 
возвращают ответ: выражение истинное или ложное.

Например, равенство проверяется оператором ==. Не следует путать его
с присваиванием =.

В выражении x = 5 применён оператор *присваивания* =: переменной x 
присвоено значение 5.

Выражение x == 5, в котором применём логический оператов 'равно' - 
это вопрос: 'А правда ли, что x равен пяти?'.

В зависимости от того, какое значение присвоено переменной x, 
программа даст ответ (вернёт значение): True (если истина) или False
(если ложь).

Ещё числа можно сравнивать операторами <, >, <=, >=, !=

Сохраним логическое выражение в переменную check и выведем её значение
на экран:
"""
check = (2 * 2 > 6)
print(check)

check = (2 * 2 == 4)
print(check)
```

```python
"""
Ветвления позволяют писать код, который исполняется, когда логическое
выражение истинно.

Вествение объявляют оператором if. После if пишут *условие* - 
логическое выражение, результатом которого может быть True или False.

Если выражение в условии истинно, то выполняется код, написанный после 
двоеточия. Если ложно, код после условия не сработает:

if <условие>:
    <код, который выполнится, если условие вернуло True>
"""
```

```python
"""
Для примера проанализируем скорость ветра по двенадцатибалльной 
шкале Бофорта. Будем менять значение переменной beaufort, и посмотрим,
как среагирует программа:
"""
beaufort = 0
if beaufort == 0:
    print('Штиль')
# если переменная не равна 0, то print не выполнится, не произойдёт ничего
```

```python
"""
Оператор if может включать блок кода, который выполнится в любом случае,
если условие ложно. Для этого существует конструкция if/else.

if <условие>:
    <код, который выполнится, если условие вернуло True>
else:
    <код, который выполнится, если условие вернуло False>

If может существовать без else, но else без if не может

Продолжим анализ скорости ветра:
"""
beaufort = 5
if beaufort == 0:
    print('Штиль')
else:
    print('Есть ветер')
```

```python
"""
Задача 1.
Научить Анфису выводить фразу: 'Новых сообщений: {значение переменной}'
Пусть переменная называется messages_count.
Пусть цикл генерирует числа от 0 до 5 и складывает эти значения в 
переменную messages_count.
Если сообщений больше 0, то должна выводиться фраза, в случае 0 пока 
ничего не делаем.
"""
for messages_count in range(6):
    if messages_count > 0:
        print('Новых сообщений: ' + str(messages_count))
```

```python
"""
Задача 2.
Расширить предыдущую программу. Если 0, то пусть выводится фраза:
'У вас нет сообщений'
"""
for messages_count in range(6):
    if messages_count > 0:
        print('Новых сообщений: ' + str(messages_count))
    else:
        print('У вас нет сообщений')
```
