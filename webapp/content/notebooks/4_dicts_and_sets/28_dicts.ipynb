```python
###############
### Словари ###
###############

"""
Часто возникают задачи, когда данные нужно хранить в виде пар, а не 
отдельными коллекциями. Например, нужно завести словарь русских 
слов и переводов к ним. С одной стороны, можно завести два отдельных 
списка: 
russian_words = ['рука', 'нога', 'хвост']
english_words = ['hand', 'leg', 'tail']

Но так хранить подобные данные неудобно. Лучше было бы хранить списки
не отдельности, а парами - русское слово и перевод:
рука - hand,
нога - leg,
хвост - tail.

В Python есть тип данных *словарь*, *dict* (от англ. dictionary - словарь).
В словарях информация хранится именно в такой структуре.
"""
```

```python
"""
Словарь - это тоже коллекция, набор элементов.

Каждый элемент словаря состоит из двух частей, между этими частями 
ставится двоеточие: первая часть элемента называется *ключ*, а вторая - 
*значение*.

Элементы словаря заключают в фигурные скобки {...}, а между собой элементы
разделяют запятой.  
"""
english = {
    'рука': 'hand',
    'нога': 'leg',
    'хвост': 'tail'
}
"""
В словаре english русские слова - это ключи, а их переводы - значения.
"""
```

```python
"""
Доступ к значению словаря по ключу.

Чтобы получить из словаря определенное значение - в коде указывают ключ,
значение которого нужно получить. Ключ указывают в квадратных скобках.
Это называется "доступ по ключу".

Приведём примеры для словаря english:
"""
english = {
    'рука': 'hand',
    'нога': 'leg',
    'хвост': 'tail'
}

print(english['рука'])
print(english['нога'])
print(english['хвост'])
```

```python
"""
Уникальность ключей в словаре.

В словаре не может быть двух одинаковых ключей. Именно это свойство 
словаря позволяет безошибочно получать доступ к каждому значению: ключ - 
это уникальный адрес для доступа к значению.

Значением в словаре может быть вообще что угодно: числа, строки, булевы
значения (True/False), списки или даже другие словари. Значения словаря 
не обязаны быть уникальными и могу повторяться.
"""
dump = {
    1: 'единица',          # Ключ - число, значение - строка.
    'земляника': 'ягода',  # Ключ - строка, значение - строка.
    'помидор': 'ягода',    # Значение 'ягода' неуникально. Так можно.
    False: 0,              # Ключ - булево значение, значение - число. 
    'лук': ['овощ', 'оружие'], # Ключ - строка, значение - список. 
    # Ключ - строка, значение - словарь. Так тоже можно!
    'англо-русский словарь': {'рука': 'hand',
                              'нога': 'leg',
                              'хвост': 'tail'
                             }
}
print(dump['лук'])
# Будет напечатано ['овощ', 'оружие']
```

```python
"""
Управление словарём.

В любом элементе словаря можно заменить существующее значение на новое - 
так же, как из ячейки шкафа можно выбросить старое содержимое и положить
туда новое.

Для замены значения нужно обратиться к элементу по ключу и присвоить ему
новое значение. В общем виде присвоение нового значения будет выглядеть
так: имя_словаря[ключ] = новое_значение

Приведём пример. В исходном словаре ключу 'рука' соответсвует значение 
'hand' - заменим его на 'arm': 
"""
english = {
    'рука': 'hand',
    'нога': 'leg',
    'хвост': 'tail'
}

english['рука'] = 'arm'

print(english['рука'])
```

```python
"""
Получение значений словаря.

Все значения словаря можно собрать в одну коллекцию. Для этого у словаря
есть метод values():
"""
english = {
    'рука': 'hand',
    'нога': 'leg',
    'хвост': 'tail'
}

# Напечаем сразу всю коллекцию
print(english.values())

# Пройдёмся по коллекции циклом
for value in english.values():
    print(value)
```

```python
"""
Получение ключей словаря.

Все ключи словаря можно собрать в одну коллекцию. Для этого у словаря
есть метод keys():
"""
english = {
    'рука': 'hand',
    'нога': 'leg',
    'хвост': 'tail'
}

# Напечаем сразу всю коллекцию
print(english.keys())

# Пройдёмся по коллекции циклом
for value in english.keys():
    print(value)
```

```python
"""
Преобразование словаря в список или сет.

Коллекции ключей и значений можно преобразовать в список или сет, применив
известные нам функции list() и set()
"""
english = {
    'рука': 'hand',
    'нога': 'leg',
    'хвост': 'tail'
}
print(list(english.keys()))
print(set(english.values()))

"""
Кроме того, преобразовать можно весь словарь целиком. Но в таком случае
множество или список будут содержать только *ключи* словаря, а значения
будут проигнорированы:
"""
# Здесь будут только ключи
print(list(english))
```

```python
"""
Создание пустых словарей и их заполнение.

Можно создать пустой словарь, а позднее наполнить его содержимым.
Создание нового элемента словаря выглядит точно так же, как и редактирование
значения: имя_словаря[ключ] = новое_значение.

Если элемент с таким ключом уже существует - то будет произведена замена 
его значения. А если нет - будет создан новый элемент словаря.

Приведем пример функции, которая добавляет в словарь название песни 
и её рейтинг:
"""
songs_and_ratings = {}

def add_song(song_name, rank):
    songs_and_ratings[song_name] = rank

# Убедимся, что словарь пока пуст:
print(songs_and_ratings)

add_song('Nothing really matters', 5)

# После вызова функции словарь пополнился:
print(songs_and_ratings)
```

```python
"""
Задача 1.
Создать словарь, где ключами будут имена людей, а значениями города, где
они живут. Путь в словаре будет 5 элементов. Затем нужно вывести на печать
город, где живёт Евгений (он должен быть одним из людей).   
"""
people_cities = {
    'Евгений': 'Москва',
    'Лиза': 'Санкт-Петербург',
    'Иван': 'Казань',
    'Диана': 'Екатеринбург',
    'Сергей': 'Новосибирск'
}

print(f'Евгений живет в городе: {people_cities["Евгений"]}')
```

```python
"""
Задача 2.
Иван переехал в другой город. Нужно заменить для него значение на другое
и напечатать фразу "Иван теперь живёт в городе ... "
(Изначальный словарь не трогать!)   
"""
people_cities = {
    'Евгений': 'Москва',
    'Лиза': 'Санкт-Петербург',
    'Иван': 'Казань',
    'Диана': 'Екатеринбург',
    'Сергей': 'Новосибирск'
}

people_cities['Иван'] = 'Волгоград'
print(f"Иван теперь живёт в городе {people_cities['Иван']}.")

```

```python
"""
Задача 3.
Количество людей в словаре прибавилось. Некоторые из них живут в одном
и том же городе. Нужно вернуть список городов, в которых они живут,
но в этом списке города не должны повторяться.

Вывод программы должен быть таким:

Мои друзья живут в следующих городах:
1. Москва
2. Санкт-Петербург
3. Казань
4. Екатеринбург
5. Новосибирск
"""
people_cities = {
    'Евгений': 'Москва',
    'Михаил': 'Москва',
    'Лиза': 'Санкт-Петербург',
    'Иван': 'Казань',
    'Диана': 'Екатеринбург',
    'Светлана': 'Екатеринбург',
    'Сергей': 'Новосибирск',
    'Тимофей': 'Санкт-Петербург'
}

unique_cities = set(people_cities.values())

print(f'Мои друзья живут в следующих городах:')
number = 0
for city in unique_cities:
    number += 1
    print(f'{number}. {city}')
```
