```python
##########################
### Расширение словаря ###
##########################

"""
Когда возникает необходимость пополнения словаря, это можно сделать 
несколькими способами. Первый способ мы уже рассмотрели - это 
добавление одного элемента. 

В этом случае мы объявляем новый ключ и присваиваем ему значение.
Но в таком случае мы, во-первых, можем добавить только одно значение,
а во-вторых мы сталкиваемся с проблемой того, что, если элемент с таким
ключом уже существует, то будет заменено его значение. И сколько бы раз
мы не меняли значение одного и того же ключа, сам он останется неизменным,
а меняться будут его значения: 
"""
english = {
    'рука': 'hand',
    'нога': 'leg',
    'хвост': 'tail'
}
# Создаём новый элемент словаря через доступ по клучу
english['голова'] = 'head'

print(english)

# Создаём несколько элементов словаря с одинаковым ключом
english['шея'] = 'neck'    # будет создан элемент с ключом 'шея'
english['шея'] = 'throat'  # значение под ключом 'шея' будет заменено

print(english)
```

```python
"""
Одновременное добавление нескольких элементов

У словарей есть метод update(), он позволяет объединить два словаря,
то есть добавить в один словарь элементы другого:
"""
english = {
    'рука': 'hand',
    'нога': 'leg',
    'хвост': 'tail'
}

# Объявим новый словарь
new_words = {'мозг': 'brain', 'палец': 'finger'}

# Добавим в словарь english элементы словаря new_words
english.update(new_words)

# Посмотрим, что теперь хранится в словаре english
print(english)

# Заодно выясним, что произошло со словарём new_words
print(new_words)
```

```python
"""
Задача 0.
Создать словарь frieds, содержащий имена трёх друзей и города, в
воторых они живут. Распечатать этот словарь.
"""
friends = {
    'Евгений': 'Москва',
    'Михаил': 'Москва',
    'Иван': 'Казань',
}
print(friends)
```

```python
"""
Задача 1.
Пополнить словарь friens ещё двумя друзьями, просто дописав их в 
объявление словаря. Распечатать результат.
"""
friends = {
    'Евгений': 'Москва',
    'Михаил': 'Москва',
    'Иван': 'Казань',
    'Сергей': 'Новосибирск',
    'Тимофей': 'Санкт-Петербург'
}
print(friends)
```

```python
"""
Задача 2.
Пополнить словарь friens ещё одним другом. На этот раз пополнить 
посредством доступа по ключу. Распечатать результат.
"""
friends = {
    'Евгений': 'Москва',
    'Михаил': 'Москва',
    'Иван': 'Казань',
    'Сергей': 'Новосибирск',
    'Тимофей': 'Санкт-Петербург'
}
friends['Лиза'] = 'Санкт-Петербург'
print(friends)
```

```python
"""
Задача 3.
Пополнить словарь friens ещё тремя друзьями, но теперь нужно создать
ещё один словарь и пополнить friends всеми его элементами за одну
операцию. Распечатать результат.
"""
friends = {
    'Евгений': 'Москва',
    'Михаил': 'Москва',
    'Иван': 'Казань',
    'Сергей': 'Новосибирск',
    'Тимофей': 'Санкт-Петербург',
    'Лиза':'Санкт-Петербург'
}

new_friends = {
    'Диана': 'Екатеринбург',
    'Ярослав':'Волгоград',
    'Алексей': 'Краснодар'
}

friends.update(new_friends)

print(friends)
```
