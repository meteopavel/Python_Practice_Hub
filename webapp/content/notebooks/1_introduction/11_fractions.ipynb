```python
#####################
### Дробные числа ###
#####################

"""
Числа бывают целыми и дробными. Для десятичных дробей (иначе их
называют "числа с плавающей запятой (или точкой)") в Python есть
специальный тип данных — float.
"""
first = 87.2
second = 50.2
third = 50.242
print(first + second + third) # Будет напечатано: 187.642
```

```python
"""
Дробные числа преобразуют в строки так же, как и целые — функцией str():
"""
first = 87.2
second = 50.2
third = 50.242
print(str(first) + str(second) + str(third))

# Будет напечатано: 87.250.250.242
# Это может быть ip-адрес одного из серверов, с которого нужно получить
# нужные нам данные
```

```python
"""
Можно преобразовать строку в дробное число, для этого есть
функция float():
"""
first = '87.2'
second = '50.2'
third = '50.242'
print(float(first) + float(second) + float(third))

# Из строк мы получили число. Будет напечатоно 187.642
```

```python
"""
Дробные числа приводят к целым функцией int(). При этом int()
*не округляет* числа по правилам арифметики, а просто отбрасывает
дробную часть.
"""
a = int(3.68)
print(a, type(a)) # Будет напечатано: 3


b = 3.25
b_int = int(3.25) # Далее в print() будет напечатано: 3
b_round = round(b, 0) # Далее в print() будет напечатано: 3.0
print(b_int, b_round, type(b_round)) # Функция type() вернёт тип

# Функция int() одинаково работает с положительными и отрицательными
# числами
c = int(-3.25)
print(c) # Будет напечатано: -3

# Можно сделать несколько преобразований в одной строке:
d = 1.5
print('Целая часть = ' + str(int(d)))
# Вернётся строка, представляющая собой целочисленную часть дроби
# Нужно обратить внимание на скобки. Количество открывающих и закрывающих
# скобок должно быть одинаково. Для удобства IDE подсвечивает 
# соответствующие скобки разными цветами
```

```python
"""
Для операций с целыми и дробными числами приводить их к одному типу
не нужно. В этом случае Python сам приведёт результат к типу float.
И если в конкретной ситуации это нас устраивает - можно так и оставить 
"""
a = 16 * 2.2 + 7 - 0.2
print(a, type(a))
# Будет напечатано 42.0 
# Ноль и точка указывают на то, что тип вернувшегося значения - float.

b = '42.0'
print(b, type(b))
# Следует обратить внимание, что строка с значением '42.0' напечатается 
# в консоли точно так же, как и дробь 42.0. В реальных проектах почти
# всегда требуется проверка типа полученных данных
# Будет напечатано 42.0

c = 16 * 2
print(c, type(c))
# Оператор * не требует приведения результата операции в типу float
# Будет напечатано 32

d = 16 / 2
print(d, type(d))
# Оператор / требует приведения результата операции в типу float
# Будет напечатано 8.0
```

```python
"""
Задача 1: Скорость света равна 1 079 252 848,8 километров в час.
Научите Анфису переводить скорость в километры в секунду.

На основе переменной speed_kmh (скорость света в км/ч) вычислите
значение переменной speed_kms (скорость света в км/с).

Значение переменной speed_kms приведите к целому типу: отбросьте
дробную часть.

В консоли напечатать: Скорость света равна <скорость света в км/с> км/с
"""
speed_kmh = 1079252848.8
speed_kms = int(speed_kmh * 3600)

print('Скорость света равна', speed_kms, 'км/с')
```

```python
"""
Задача 2: Длина Питона из известного советского мультфильма равна 38 попугаев
и одно попугайское крылышко. Будем считать, что крылышко равно 0.2
попугая.

В железнодорожный вагон можно уложить по длине шесть с половиной
Питонов. Помогите Анфисе выяснить, какое количество попугаев можно
уложить вдоль вагона.

Попугаи бывают только целые, это важно. Поэтому переменная result
должна быть типа int.

В консоли напечатать: В вагоне можно положить в ряд <result> попугаев
"""
snake = '38.2'
length = 6.5

result = int(float(snake) * length)

print('В вагоне можно положить в ряд', result, 'попугаев')
```
