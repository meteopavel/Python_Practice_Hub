```python
####################
###  Переменные  ###
####################

"""
Программы, которые не могут сохранять данные практически бесполезны. 

*Переменная* - работает как подписанная коробка или как помеченная
ячейка, куда можно что-то положить и не потерять.

Когда в коде впервые пишут имя переменной, это называется
*объявление переменной*. Имя придумывает сам программист.
Однако следует помнить, что в программировании очень важен вопрос
именования сущностей, в том числе и переменных. Следует называть их
согласно их содержания, чтобы по названию было понятно, что содержится
в переменной.

Для объявления переменной достаточно просто указать имя, поставить
знак "равно" = (этот знак в Питоне называется *оператор присваивания*)
и написать значение, которое будет храниться в переменной.
"""
x = 1     # теперь в переменной x хранится значение 1
print(x)  # убедимся в этом, передав x в качестве аргумента print()
x = 38    # значение можно изменить после объявления
print(x)  # убедимся в этом
```

```python
"""
В переменных можно сохранить не только числа: в них можно "положить"
и строку. 
"""
message = 'Привет, Мир!' # объявили переменную message и присвоили ей 
                         # значение в виде строки 'Привет, Мир!'
"""
Когда переменная объявлена и ей присвоено значение - имя переменной
можно указать в коде, и вместо имени будет подставляться её значение.
"""
print(message) # В функцию передан не текст, а переменная, в которой 
               # хранится текст.
```

```python
#####################################
###  Типы данных: числа и строки  ###
#####################################

"""
Значения переменных различаются по своей сути: например, значения могут
быть *числами* или *строками*. Это разные *типы данных*. В Python есть и
другие типы, но для начала хватит и этих двух.

Строка записывается как символ или набор символов внутри одинарных или
двойных кавычек:
"""
genre = 'Роман'               # строка в одинарных кавычках
author = "Фёдор Достоевский"  # строка в двойных кавычках
title = 'Идиот'
year = '1868'                 # число в кавычках становится строкой
space = ' '                   # даже пробел является строкой
empty = ''                    # пустая строка - это тоже строка

"""
Для обрамления строк можно пользоваться одинарными или двойными кавычками,
но главное - открывающая и закрывающая кавычки должны быть одинаковыми.
"""
# author = "Марк Твен'        # ошибка 
"""
Традиционно в проектах используются одинарные кавычки
"""
```

```python
"""
Иногда в проектах встречается необходимость выводить символы кавычек
в консоль. Это может приводить к проблемам. Приведём несколько примеров:
"""
# При использовании кавычек разного типа проблемы не будет:
print('Это опасное явление погоды называется "шквал"')
# Здесь тоже не будет проблемы, НО теперь мы использовали
# кавычки, которые отличаются от традиционных (одинарных).
# И это будет нарушать единый принцип использования кавычек:
print("Это опасное явление погоды называется 'шквал'")
# В некоторых случаях проблему можно решить экранированием.
# Оно делается путём подстановки символа \ перед экранируемым
# символом. Это означает, что Python не будет воспринимать 
# экранируемый символ как значимый для кода. Это будет просто текст.
print('Это опасное явление погоды называется \'шквал\'')
```

```python
"""
Числовые значения записываются без кавычек:
"""
twenty_five = 25
about_pi = 3
"""
В редакторе кода числа и строки выделяются разными цветами, чтобы
читать исходный код было легче.
"""
```

```python
"""
Со значениями разных типов Python работает по-разному.
С переменными, которые содержат числа, можно проводить любые
арифметические операции: например, складывать их при помощи оператора +:
"""
twenty_five = 25
about_pi = 3
# Переменные одного типа можно складывать,
# а получившееся значение присваивать другой переменной
total = twenty_five + about_pi
print(total)
```

```python
"""
Но если те же значения переменных записать как строки (взять значения
в кавычки) — результат изменится.
"""
twenty_five = '25'
about_pi = '3'
# При сложении двух строк Python "склеивает" содержимое этих строк в
# одну.
total = twenty_five + about_pi
print(total)
# Объединение нескольких строк в одну называют "конкатенацией"
# (от лат. concatenatio, "присоединение, сцепление").
```

```python
"""
Можно складывать числа с числами — это арифметика. Можно складывать
строки со строками — это конкатенация.
А вот если складывать числа со строками — получится полная ерунда
и ошибка в программе. Складывать можно значения одинакового типа.
"""
num_as_string = '100499'
num_as_num = 1

print(num_as_string + num_as_num)
# Этот пример выдаст ошибку, потому что используемые переменные
# разных типов. Нельзя конкатенировать строку и число
```

```python
"""
Задача. Анфису нужно поселить в фитнес-трекер. Нужно научить её
подсчитывать количество шагов пользователя и сообщать ему результат.

В переменных steps_today и steps_yesterday записано, сколько шагов
прошёл незнакомый нам Геннадий вчера и сегодня. 

Нужно вычислить в коде сумму шагов за два дня и сохранить результат
в переменную steps_sum. А программа напечатает получившийся результат.
"""
steps_yesterday = 8452
steps_today = 6783

# Вычислить количество шагов за два дня
steps_sum = steps_yesterday + steps_today

print('Сколько шагов сделал Геннадий за два дня?')
# Вывести количество шагов вторым принтом
print(steps_sum)
```
