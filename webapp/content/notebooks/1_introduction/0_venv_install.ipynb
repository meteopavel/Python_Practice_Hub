```python
"""
Некоторые команды для работы с виртуальным окружением (venv)
Команды нужно выполнять в консоли GitBash, так как эта консоль
может выполнять команды ядра Linux

py -0 (только для Windows) - выводит список всех версий python
в системе

python --version - выводит версию python, которая является основной
для Windows (по умолчанию)

py -3.11 -m venv venv - установит виртуальную среду python (версия 3.11)
в папку venv относительно того пути, где сейчас находится консоль

python -m venv venv - установит виртуальную среду python, который
является основным для системы

source venv/Scripts/activate - активирует виртуальную среду. При
активированной venv внизу после всех выводов консоли всегда должна
появлятся надпись (venv). В среде Linux эта команда отличается

pip list - выводит список пакетов, установленных в текущей среде.
Кроме того, этой командой удобно проверять, в какой среде мы сейчас
работаем

python -m pip install --upgrade pip  - обновляет пакетный менеджер

pip install <имя пакета>==<версия> - устанавливает пакет в активированную 
среду (не забыть проверить, активирован ли venv)

pip uninstall <имя пакета> - удаляет пакет из среды

pip freeze -> requirements.txt - сохраняет текущие установленные пакеты
в текстовый файл

pip install -r requirements.txt - устанавливает пакеты, прописанные 
в файле

Если работаем с файлами ipynb (jupyter), то нужно установить ipykernel
в venv (VS Code предложит это сделать - лучше не соглашаться). Вместо
этого делаем requirements.txt. Прописываем туда ipykernel. Можно прописать
без указания версии. Тогда будет установлена самая новая

deactivate - деактивирует виртуальную среду
"""
```
