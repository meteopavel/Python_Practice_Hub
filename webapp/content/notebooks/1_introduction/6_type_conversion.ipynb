```python
############################
### Преобразование типов ###
############################

"""
Со значениями разных типов операторы Python работают по-разному:
оператор + арифметически складывает числа и в то же время 
"склеивает" строки. 
"""
one_hundred = 100
five_hundred = 500
print(one_hundred + five_hundred)
# Будет напечатано: 600

ha = 'ха'
print(ha + ha + ha + ha)
# Будет напечатано: хахахаха 
```

```python
"""
Строку можно умножить на целое число (оператор умножения — звёздочка *)
"""
# Если умножить число на число - никаких неожиданностей не будет
six = 6
print(six * 7)

# А умножение строки на число скопирует строку несколько раз
ho = 'хо'
print(ho * 10)
```

```python
"""
Такой код не сработает, Python выдаст ошибку:
"""
# Объявили две переменные разных типов
number = 100
rubles = ' рублей'

print(number + rubles)
# В тексте ошибки будет сказано, что оператор '+' 
# не складывает целые числа со строками
```

```python
"""
Одно из решений этой проблемы - привести число к строке (конвертировать
тип данных)

Преобразованием данных из одного типа в другой занимаются специальные
функции. Так, для превращения числа в строку вызывают функцию str():
"""
number = 100
rubles = ' рублей'

print(str(number) + rubles)
# Код, где число преобразовано в строку ("приведено к строке"),
# прекрасно работает
```

```python
"""
А преобразовать строку в целое число можно функцией int():
"""
# Преобразуем строки в числа и сложим
one_hundred = '100'
five_hundred = '500'
print(int(one_hundred) + int(five_hundred))
```

```python
"""
Задание:
Научим Анфису информировать вас о новых сообщениях, которые вы могли
бы получить.
Выведите на экран строку 'У вас 8 новых сообщений', составленную
* из строки 'У вас ',
* значения переменной count
* и из строки ' новых сообщений'.
Само сообщение упаковать в переменную и уже её распечатать
"""
count = 8
message = 'У вас ' + str(count) + ' новых сообщений'
print(message)
```
