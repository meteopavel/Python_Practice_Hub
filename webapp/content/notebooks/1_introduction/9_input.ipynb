```python
#######################
### Функция input() ###
#######################

"""
Функция input() в Python используется для получения данных от
пользователя через стандартный ввод (обычно это клавиатура).
Она позволяет программе взаимодействовать с пользователем,
запрашивая у него ввод информации.
"""
```

```python
"""
input() может принимать один необязательный аргумент promt. 
Синтаксически это будет выглядеть так:

input([prompt]) # квадратные скобки указывают, что 
                # аргумент необязательный

promt - это строка, которая будет выведена на экран перед тем, как
программа будет ожидать ввода от пользователя. Это может быть полезно
для пояснения того, что именно нужно ввести
"""
input('Введите число: ')
```

```python
"""
Функция input() всегда возвращает введенные данные в виде
строки (тип str). Если нужно получить другой тип данных, необходимо
использовать функции преобразования (например int() или float()). 
"""
name = input('Введите ваше имя: ')
print('Привет, ' + name + '!')
# получить возраст, далее с его помощью посчитать год рождения
# вывести две фразы: 1. Вам *** лет. 2. Вы родились в *** году.
age = int(input('Введите ваш возраст: '))
birth_year = 2024 - age # хардкод не допустим в реальных проектах
print('Вам ' + str(age) + ' лет.')
print('Вы родились в ' + str(birth_year) + ' году.')
```

```python
"""
При преобразовании типов важно учитывать возможные ошибки. Например,
если пользователь введёт нечисловое значение, программа вызовет ошибку.
Для проверки можно использовать обработку исключений:
"""
try:
    age = int(input('Введите ваш возраст: '))
    print('Вам ' + str(age) + ' лет.')
except ValueError:
    print('Ошибка: введите числовое значение.')
```
