```python
##############
### Ошибки ###
##############

"""
Разработчики постоянно делают ошибки в коде. Написать код, а затем
отловить и исправить баги - это стандартный путь программиста.

Python сильно помогает найти ошибку. Он предоставляет всю возможную
информацию о проблеме. Например, "Я сломался, ошибка на строке 8, там
забыли закрыть скобку!"

Python пишет об ошибках только по-английски. Однако не нужно обладать 
глубокими знаниями, чтобы понять, о чём он пытается сообщить. Нужно
читать всё, что он пишет. Это поможет самостоятельно разбраться в
проблеме. 

Сообщения об ошибках называются traceback
(по-русски говорят "трейсбэк").

Приведём пример трейсбэка:

Traceback (most recent call last):
    File "main.py", line 1    # файл и номер строки, где ошибка
        print('Hello, world!) # сама строка, в которой ошибка
                            ^ # конкретное место, в котором ошибка

SyntaxError: EOL while scanning string literal # тип и описание ошибки
"""
```

```python
"""
Ошибка "Неизвестное имя"
"""
one_hundred = 100
print(onehundred) # здесь мы намеренно допустили опечатку
"""
Данный код выдаст трейсбэк со следующим содержанием:
* в какой строчке ошибка (line 5)
* тип ошибки (NameError)
* из-за чего произошла ошибка (name 'onehundred' is not defined)
  (никаких сущностей с именем onehundred ещё не объявлялось)
"""
```

```python
"""
В следующем примере ошибка будет только на первой строке, а вторая будет
без ошибок. При этом нужно обратить внимание, на какую строку укажет
Python
"""
print(3 + 5
print('Всё в порядке!')
"""
Начиная с версии 3.11 возникает ошибка
SyntaxError: '(' was never closed

Хотя в более ранних версиях ошибка возникнет не на первой,
а на второй строке с кодом. Она будет выглядеть так:
    print('Всё в порядке!')
    ^
SyntaxError: invalid syntax

Python разрешает ставить открывающую и закрывающую скобки на разных
строчках. Поэтому в более ранних версиях он терпеливо дожидался,
пока код совершенно утратит смысл, и только тогда выдавал ошибку.

Этот код показал и ещё одну особенность поведения Python: 
при любой ошибке выполнение кода полностью останавливается.
Код после строки с ошибкой не будет выполнен, даже если он без ошибок
"""
```

```python
from ipaddress import ip_address, ip_network

for mask in range(32, 1, -1):
    net = ip_network(f'215.181.192.0/{mask}')
    if ip_address('215.181.200.27') in net:
        print(mask, net[-1])

print(int('11110000', 2))

"""
Данный код упадёт с ошибкой не сразу, поскольку в нём нет синтаксических
или других явных ошибок, выявляемых даже на уровне IDE. Но ошибка
возникнет при обращении к функциям других библиотек на определённом 
этапе исполнения. 

До падения код будет выдавать результаты в консоль, и эти результаты 
сохранятся.
"""
```
